### 내 답안 R-1

DataFrame의 역할 =  df는 2차원 - 즉 로우와 컬럼을 표현하는 표형태로서 제시되며ㅡ 모델이 보는 값보다는 사람이 보고 정규화 또는 전처리, 기술통계, 데이터타입, 이상치/결측치 유무를 확인하고 x y를 최종점검하며 피처와 타겟간의 관계를 점검하고 보정하는데 사용한다. 
Series의 역할 =  시리즈는 행렬이라기보단 벡터에 가까우며, 특정 컬럼에 대한 값들의 집합을 의미한다. 즉 시리즈는 1D인 라벨의 ARRAY, 즉 배열이다.
ndarray의 역할 =  NDARRAY는, DF와는 다르게, NP. 기반으로 모델의 연산에 특화되어있다. DF는 사람이 보는 용도라면, ND는 모델이 보는 것으로서, 기존 DTYPE, INDEXING, LABEL, CLOUMN, FEATURE_NAME 등 전형적인 2D ARRAY에서 - 연산을 위해 인덱싱 라벨, 컬럼 등이 제거되는 형태를 취한다. 따라서ㅡ, 피처간의 순서, 타겟과의 관계 등은 남아있으나 그외의 컬럼명 피처 세부명칭 등은 삭제됨으로 사전에 각 피처와 컬럼을 정리해서 남겨둬야하며, 피처의 의미가 사라짐을 항상 염두에 두어야한다.
tensor의 역할 =  텐서란, ND와 비슷하게 모델이 계산하기 좋은 구조를 가지나, 구체적으로 MLP 등 딥러닝에 이용되며 그 이유로는 채널의 개념과 벡터 활용에 보다 유효한 성능을 가지는데, 이는 GPU등의 연산체계를 적극활용하기 때문이다.

사람이 의미를 확인하기 좋은 구조 =  DF
모델이 계산하기 좋은 구조 =  NDARRAY, TENSOR
가장 조심해야 할 혼동 =  시리즈와 DF는 다르며, SHAPE와 VALUE는 다르다. 시리즈는, Y1 = SERIES, SERIES = (N,) / Y2 = DF, SERIES = (N,1)로 둘시Y1은 특정 행에 있는 데이터의 배열이므로 1D ARRAY로 볼수 있으나 Y2는 행1에 대하여 N의 배열으로서 2D ARRAY이다.

````python
# 참고
images = np.load(DATA_DIR / "mock_images.npz")
print(images["images_gray"].shape)  # (12, 28, 28): 12개의 샘플(이미지), 각 이미지가 28x28 픽셀
print(images["images_flat"].shape)  # (12, 784): 28x28을 일렬(1차원)로 펴면 784, 이미지 12장
print(images["labels"].shape)       # (12,): 12개 이미지에 대한 레이블(타겟), 각 샘플의 정답값
#12는 class id가 아닌 sample(이미지) 수. 즉, 데이터셋 내 총 이미지 개수.  
# Flatten은 shape(차원)만 펴는 것, 값의 범위/스케일(normalization)과는 무관함
````

### 내 답안 R-4

12의 의미 =  image(x,y,z)에서 x는 axis=0 , y=axis=1, z=axis=2 를 의미하는데, 이떄 있어 12는 12개의 이미지를 의미한다.
28 x 28의 의미 =  axis 0,1 에서 각각 28까지의 길이가 있다는것. 즉 이미지에서는 가로 28 , 세로 28픽셀인것.
784의 의미 =  flatten 진행시, (28,28) 이미지의 특정 값은, 2차원 2darray로서 가로위치정보 세로위치정보, 하나의 위치에서 위 아래 대각 등 상호 위치정보를 가진다. 이떄 있어, 플래튼을 진행시, 2d->1d로 변환되는데., 이로 인해 가로로 쭉 늘여뜰여진다. 따라서 시리즈로 표현하면, series = (784,) 가 될 것ㅇ다.
labels는 무엇인가 =  타겟값 y에 해당하는 컬럼의 의미이자 이름이며, 해당에서 표현은 y.shape = (n,) 이 될 것이다
normalization과 Flatten의 차이 =  정규화는 각각의 서로 다른 피처값들의 범위를 동일한 범위내로 , 예를들어 0~1 사이로, 스케일링해주는 것을 얘기하나 플래튼은 평탄화 즉 차원을 주로 내린다. 즉, 배열을 바꾼다. 
Flatten하면 사라지는 정보 =  784의 의미에서 설명함함

### 내 답안 R-5

DataFrame 단계의 역할 =  정규화 기술통계 전처리 컬럼의 이름 피처의의미 등등
head/info/describe의 역할 =  데이터에 대하여 () 범위내 처음부터 () 까지 보여줌. / 인포는 그 데이터 타입 등을 식별하는데 유용 / 디스크라입은 기술통계량 제시
X/y 분리의 이유 =  모델에 들어갈 입력과 모델이 출력할 출력값을 정의하고 타겟과 피처를 나눔
categorical/결측 처리를 DataFrame 단계에서 하는 이유 =  범주화에 대해서 결측처리는, 엔디어레이로 들어갈시 컬럼의 의미가 사라짐으로서 피처의 의미를 직접 보고 확인하기 어려우며, 데이터타입이 숫자가 아닌 범주형 str 등일 경우 입력 자체가 불가하므로 오염됨
ndarray/tensor로 바꾸는 이유 = 모델용계산  
model.fit(X_train, y_train)의 의미 =  데이터를 x에 대한 피처로서 모델이 이를 확인 및 학습하여 출력 y를 도출시키도록 한다는것. 

최종 한 문장 요약 =  모델링.
