## 3. 정수계획 핵심 기준표
| 구분 | 변수 의미 | domain | 대표 제약 |
|---|---|---|---|
| pure integer | 모든 의사결정이 정수 수량 | `x_j` integer | 개수, 대수, 인원 수 제약 |
| mixed integer | 일부는 연속 수량, 일부는 정수/0-1 | `x_j >= 0`, `y_j in {0,1}` | `x_j <= U_j y_j` |
| binary integer | 선택 여부 자체 | `x_j in {0,1}` | covering, budget, prerequisite |
| LP relaxation | 정수조건을 제거한 완화 문제 | `0 <= x_j <= 1` 또는 연속 정수변수 | bound 산출용이며 반올림 답안 금지 |
| fixed-charge | 가동 여부와 생산량 분리 | `x_j >= 0`, `y_j in {0,1}` | `L_j y_j <= x_j <= U_j y_j` |
