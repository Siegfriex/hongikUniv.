## 1. 핵심 기준표

| 기준 | 판별법 |
| --- | --- |
| `<=` 제약 | slack variable을 더한다. |
| `>=` 제약 | surplus variable을 빼고 artificial variable을 더한다. |
| `=` 제약 | 초기 basis가 없으면 artificial variable을 더한다. |
| Phase I | artificial variable 합 `w`를 최소화한다. |
| Phase II | `w*=0`이면 artificial을 제거하고 원 목적함수로 최적화한다. |
