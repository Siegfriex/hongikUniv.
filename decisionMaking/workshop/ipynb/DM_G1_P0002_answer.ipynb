## 2.5 시각화 심화 학습 보드

이 섹션은 P0002의 two-phase 구조를 시각적으로 정리한다. 핵심은 다음이다.

1. **부등호 방향이 보조변수의 부호를 결정한다.**
2. **artificial variable은 원문제 변수가 아니라 출발 basis를 만들기 위한 임시 장치다.**
3. **Phase I의 `w=0`은 feasibility 신호이지 원 목적함수 값이 아니다.**


```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch, Polygon

# Matplotlib에 확실히 존재하는 DejaVu Sans를 사용한다.
# 그림 내부 텍스트는 폰트 깨짐 방지를 위해 ASCII/English로 제한하고,
# 한국어 해설은 Markdown 셀에서 제공한다.
mpl.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.unicode_minus": False,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 10,
})

def box(ax, xy, wh, text, fc="#eef6ff", ec="#243238", fs=9):
    x, y = xy
    w, h = wh
    ax.add_patch(Rectangle((x, y), w, h, facecolor=fc, edgecolor=ec, lw=1.4))
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", fontsize=fs)

def arrow(ax, start, end, color="#2563eb"):
    ax.add_patch(FancyArrowPatch(start, end, arrowstyle="-|>", mutation_scale=14, lw=1.6, color=color))

def draw_flow(ax, rows, title):
    ax.axis("off")
    ax.set_title(title)
    n = len(rows)
    colors = ["#e7f0fd", "#e9f7ef", "#fff4d6", "#fce8e6", "#ece7fb", "#e8f6f8"]
    centers = []
    for i, row in enumerate(rows):
        y = 0.90 - i * (0.78 / max(n-1, 1))
        row_centers = []
        m = len(row)
        for j, txt in enumerate(row):
            x = 0.08 + j * (0.84 / max(m-1, 1)) if m > 1 else 0.38
            box(ax, (x, y-0.055), (0.24, 0.11), txt, fc=colors[(i+j) % len(colors)], fs=8.5)
            row_centers.append((x+0.12, y))
        centers.append(row_centers)
    for i in range(len(centers)-1):
        for a in centers[i]:
            for b in centers[i+1]:
                if len(centers[i]) == 1 or len(centers[i+1]) == 1 or abs(a[0]-b[0]) < 0.28:
                    arrow(ax, (a[0], a[1]-0.06), (b[0], b[1]+0.06))

print("Visualization setup ready. Matplotlib font:", mpl.rcParams["font.family"][0])
```

```python
# P0002 Board 1: constraint type conversion and why artificial variables appear
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

draw_flow(
    axes[0],
    [
        ["<= constraint"],
        ["add slack +s"],
        ["basis column available"],
        [">= constraint"],
        ["subtract surplus -s"],
        ["add artificial +a"],
        ["= constraint"],
        ["add artificial +a"]
    ],
    "Constraint type -> standard form"
)

ax = axes[1]
ax.axis("off")
ax.set_title("Surplus sign intuition")
box(ax, (0.05, 0.72), (0.25, 0.12), "LHS >= RHS", fc="#eef6ff")
box(ax, (0.38, 0.72), (0.25, 0.12), "LHS = RHS + excess", fc="#fff4d6")
box(ax, (0.70, 0.72), (0.25, 0.12), "LHS - surplus = RHS", fc="#e9f7ef")
arrow(ax, (0.30,0.78), (0.38,0.78)); arrow(ax, (0.63,0.78), (0.70,0.78))
box(ax, (0.12, 0.42), (0.32, 0.14), "surplus alone:\nno +1 basis column", fc="#fce8e6")
box(ax, (0.56, 0.42), (0.32, 0.14), "add artificial:\ninitial basis exists", fc="#ece7fb")
arrow(ax, (0.44,0.49), (0.56,0.49))
box(ax, (0.32, 0.13), (0.36, 0.13), "Phase I removes artificial", fc="#e8f6f8")
arrow(ax, (0.72,0.42), (0.50,0.26))
plt.tight_layout()
plt.show()
```

```python
# P0002 Board 2: Phase I and Phase II are different questions
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

draw_flow(
    axes[0],
    [
        ["Original constraints"],
        ["Standard form"],
        ["Artificial basis"],
        ["Phase I: min w=sum(a)"],
        ["w*=0?"],
        ["Remove artificial columns"],
        ["Phase II: optimize original z"]
    ],
    "Two-phase method pipeline"
)

ax = axes[1]
ax.set_title("Do not confuse w=0 with z=0")
names = ["Phase I w", "Original z"]
values = [0, 196]
bars = ax.bar(names, values, color=["#60a5fa", "#fb923c"])
ax.text(0, 8, "feasible basis found", ha="center", fontsize=9)
ax.text(1, 160, "optimized only in Phase II", ha="center", fontsize=9)
ax.set_ylabel("different objective values")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# P0002 Board 3: diet-style matrix and Solver mapping
fig, axes = plt.subplots(1, 2, figsize=(15, 5.8))

ax = axes[0]
mat = np.array([[10, 20, 10], [0, 10, 30]], dtype=float)
im = ax.imshow(mat, cmap="Blues")
ax.set_xticks(range(3)); ax.set_xticklabels(["Food1", "Food2", "Food3"])
ax.set_yticks(range(2)); ax.set_yticklabels(["Vitamin A", "Vitamin C"])
for i in range(mat.shape[0]):
    for j in range(mat.shape[1]):
        ax.text(j, i, f"{mat[i,j]:g}", ha="center", va="center", fontsize=11)
ax.set_title("Diet matrix: variables feed constraints")
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

draw_flow(
    axes[1],
    [
        ["Food amount cells"],
        ["Cost target cell"],
        ["Nutrient LHS cells"],
        [">= requirement RHS"],
        ["Solver finds feasible optimum"],
        ["Manual two-phase explains the basis"]
    ],
    "Diet problem -> Solver mapping"
)
plt.tight_layout()
plt.show()
```

## 4. 오답튜터 기준표

| 오류 | 피드백 기준 |
| --- | --- |
| artificial variable을 실제 변수로 해석 | 임시 계산 변수이며 최종적으로 0이어야 함을 설명한다. |
| surplus 부호를 +로 씀 | `>=` 제약의 초과량은 좌변에서 빼야 함을 현실 언어로 설명한다. |
| Phase I objective를 원문제 objective로 착각 | `w`와 `z`의 목적을 분리한다. |
| w=0과 z=0 혼동 | `w=0`은 feasible basis 존재만 뜻한다고 교정한다. |
| artificial 양수인데 feasible 판정 | 원 제약만으로는 feasible basis가 없다는 신호라고 교정한다. |
