```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch, Polygon

# Matplotlib에 확실히 존재하는 DejaVu Sans를 사용한다.
# 그림 내부 텍스트는 폰트 깨짐 방지를 위해 ASCII/English로 제한하고,
# 한국어 해설은 Markdown 셀에서 제공한다.
mpl.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.unicode_minus": False,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 10,
})

def box(ax, xy, wh, text, fc="#eef6ff", ec="#243238", fs=9):
    x, y = xy
    w, h = wh
    ax.add_patch(Rectangle((x, y), w, h, facecolor=fc, edgecolor=ec, lw=1.4))
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", fontsize=fs)

def arrow(ax, start, end, color="#2563eb"):
    ax.add_patch(FancyArrowPatch(start, end, arrowstyle="-|>", mutation_scale=14, lw=1.6, color=color))

def draw_flow(ax, rows, title):
    ax.axis("off")
    ax.set_title(title)
    n = len(rows)
    colors = ["#e7f0fd", "#e9f7ef", "#fff4d6", "#fce8e6", "#ece7fb", "#e8f6f8"]
    centers = []
    for i, row in enumerate(rows):
        y = 0.90 - i * (0.78 / max(n-1, 1))
        row_centers = []
        m = len(row)
        for j, txt in enumerate(row):
            x = 0.08 + j * (0.84 / max(m-1, 1)) if m > 1 else 0.38
            box(ax, (x, y-0.055), (0.24, 0.11), txt, fc=colors[(i+j) % len(colors)], fs=8.5)
            row_centers.append((x+0.12, y))
        centers.append(row_centers)
    for i in range(len(centers)-1):
        for a in centers[i]:
            for b in centers[i+1]:
                if len(centers[i]) == 1 or len(centers[i+1]) == 1 or abs(a[0]-b[0]) < 0.28:
                    arrow(ax, (a[0], a[1]-0.06), (b[0], b[1]+0.06))

print("Visualization setup ready. Matplotlib font:", mpl.rcParams["font.family"][0])
```

```python
# P0001 Board 1: feasible region, BFS movement, and alternate-optimal face
def feasible_grid_p1():
    x = np.linspace(0, 70, 500)
    y = np.linspace(0, 70, 500)
    X, Y = np.meshgrid(x, y)
    F = (2*X + Y <= 120) & (X + 2*Y <= 100) & (X <= 45) & (X >= 0) & (Y >= 0)
    return X, Y, F

fig, axes = plt.subplots(1, 2, figsize=(14, 5.8))

X, Y, F = feasible_grid_p1()
ax = axes[0]
ax.contourf(X, Y, F.astype(float), levels=[0.5, 1.5], colors=["#d9f2e3"], alpha=0.95)
xs = np.linspace(0, 70, 400)
ax.plot(xs, 120 - 2*xs, label="2x1+x2<=120")
ax.plot(xs, (100 - xs)/2, label="x1+2x2<=100")
ax.axvline(45, color="#555", ls="--", label="x1<=45")
path = np.array([[0, 0], [45, 0], [45, 27.5]])
ax.plot(path[:,0], path[:,1], "o-", color="#dc2626", lw=3, label="BFS path")
for p, t in [((0,0),"start"), ((45,0),"after 1st pivot"), ((45,27.5),"optimum")]:
    ax.text(p[0]+1, p[1]+2, t, fontsize=9)
ax.set_xlim(0, 70); ax.set_ylim(0, 70)
ax.set_xlabel("x1"); ax.set_ylabel("x2")
ax.set_title("Simplex movement: BFS to adjacent BFS")
ax.grid(alpha=0.25); ax.legend(fontsize=8)

ax = axes[1]
poly = np.array([[0,0], [8,0], [8,2], [2,8], [0,8]])
ax.add_patch(Polygon(poly, closed=True, fc="#d9f2e3", ec="#245b40", alpha=0.95))
ax.plot([2,8], [8,2], color="#dc2626", lw=5, label="optimal face")
ax.scatter([4,6,5], [6,4,3], c=["#dc2626","#dc2626","#2563eb"], s=65)
ax.text(2.2, 8.25, "same objective value on this edge", fontsize=9)
ax.text(5.2, 2.75, "feasible but not optimal", fontsize=9)
ax.set_xlim(0, 10); ax.set_ylim(0, 10)
ax.set_xlabel("x1"); ax.set_ylabel("x2")
ax.set_title("Multiple optima: only points on optimal face")
ax.grid(alpha=0.25); ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
# P0001 Board 2: ratio test, reduced cost, and unbounded signal
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))

ax = axes[0]
rows = ["row1", "row2", "row3", "row4"]
ratios = [60, 100, 45, np.nan]
bars = ax.bar(rows[:3], ratios[:3], color=["#93c5fd", "#93c5fd", "#f87171"])
ax.text(2, 48, "min positive ratio\nleaving row", ha="center", fontsize=9)
ax.set_title("Minimum ratio test")
ax.set_ylabel("RHS / positive column coefficient")
ax.grid(axis="y", alpha=0.25)

ax = axes[1]
vars_ = ["x3", "x4", "x5"]
reduced = [0, 2, 1]
ax.bar(vars_, reduced, color=["#f87171", "#93c5fd", "#93c5fd"])
ax.axhline(0, color="#111", lw=1)
ax.text(0, 0.18, "zero reduced cost\nalternate candidate", ha="center", fontsize=9)
ax.set_title("Reduced cost signal")
ax.set_ylabel("nonbasic coefficient")
ax.grid(axis="y", alpha=0.25)

ax = axes[2]
ax.set_xlim(0, 10); ax.set_ylim(0, 7)
poly = np.array([[0,0], [10,0], [10,5.8], [3.2,3], [0,1.7]])
ax.add_patch(Polygon(poly, closed=True, fc="#fdebd3", ec="#92400e", alpha=0.95))
ax.arrow(2.5, 1.6, 6.6, 3.4, width=0.04, head_width=0.28, color="#dc2626", length_includes_head=True)
ax.text(5.0, 5.5, "improving ray\nno leaving row", color="#7f1d1d", fontsize=9)
ax.set_title("Unbounded: improving direction has no blocker")
ax.set_xlabel("entering variable direction")
ax.set_ylabel("other variable")
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
# P0001 Board 3: Solver mapping and termination routing
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

draw_flow(
    axes[0],
    [
        ["Decision variables"],
        ["Changing cells: x1, x2"],
        ["Target cell: profit"],
        ["Constraint LHS cells"],
        ["RHS bounds + nonnegativity"],
        ["Simplex LP"]
    ],
    "LP -> Excel Solver mapping"
)

draw_flow(
    axes[1],
    [
        ["Tableau status"],
        ["Improving coefficient?", "No improving coefficient?"],
        ["No ratio candidate", "Zero reduced cost?"],
        ["Unbounded", "Multiple optimum"],
        ["Phase I w*>0"],
        ["Infeasible"]
    ],
    "Special termination signal routing"
)
plt.tight_layout()
plt.show()
```
