## 2. 문항별 정답


### 문항 1

- 정답: `2x1+x2+s1=120`, `x1+2x2+s2=100`, `x1+s3=45`, 모든 변수는 0 이상이다. 초기 기저변수는 `s1,s2,s3`, 비기저변수는 `x1,x2`이다. 초기 BFS는 `(x1,x2,s1,s2,s3)=(0,0,120,100,45)`이다.
- Solver mapping: changing cells는 `x1,x2`, target cell은 `5x1+4x2`, constraint LHS cells는 `2x1+x2`, `x1+2x2`, `x1`, RHS cells는 `120,100,45`이다.
- 근거: `DM_PDF01:p001:L003`, node_id=`n_DM_PDF01.simplex_tableau`.
- 자주 하는 오답: slack을 빼거나 초기 기저변수에 원변수를 포함한다.
- 최종 답안형: 표준형, basis/nonbasis, 초기 BFS, Solver mapping을 분리해서 작성.


### 문항 2

- 정답: max 표준 타블로에서 목적행을 `z-5x1-4x2=0`으로 쓰면 Dantzig rule 기준 entering variable은 계수 절댓값이 더 큰 `x1`이다. ratio 후보는 `120/2=60`, `100/1=100`, `45/1=45`이고 leaving variable은 최소비율의 `s3`이다.
- 제외 행: entering column 값이 0 이하인 행이 있다면 제외한다. 이 문제에서는 `x1` column의 세 제약계수가 모두 양수라 제외 행이 없다.
- 근거: `DM_PDF01:p001:L003`, node_id=`n_DM_PDF01.entering_leaving_variable`, `n_DM_PDF01.minimum_ratio_test`.
- 자주 하는 오답: 음수 계수를 절댓값으로 바꿔 ratio 후보에 넣는다.


### 문항 3

- 정답: pivot element는 `s3` 행의 `x1` 계수 1이다. pivot row는 이미 1이므로 그대로 두고, 다른 행의 `x1` 계수를 0으로 만들기 위해 행 연산을 수행한다. 새 기저변수는 `s1,s2,x1`이다.
- 그래프 대응: 현재 꼭짓점에서 `x1` 방향으로 이동하다가 `x1<=45` 경계에 먼저 닿는 인접 꼭짓점 이동이다.
- 근거: `DM_PDF01:p001:L003`, node_id=`n_DM_PDF01.pivot_operation`.
- 자주 하는 오답: pivot row만 바꾸고 다른 행의 entering column을 소거하지 않는다.


### 문항 4

- 정답: 개선 가능한 계수가 없다는 전제라면 현재 해는 최적이다. 동시에 `x3`의 nonbasic coefficient가 0이므로 복수 최적해 가능성이 있다. `x3`는 목적값을 악화시키지 않고 기저에 들어올 수 있는 대안 entering 후보를 뜻한다.
- Solver 주의점: Solver가 한 최적해만 출력해도 대안 최적해가 존재할 수 있으므로 유일해라고 단정하면 안 된다.
- 근거: `DM_PDF01:p002:L003`, node_id=`n_DM_PDF01.optimality_test`, `n_DM_PDF01.multiple_optima`.
- 자주 하는 오답: reduced cost 0을 “변수가 필요 없다”는 뜻으로 해석한다.


### 문항 5

- 정답: 비기저변수의 목적함수 계수 0은 해당 변수를 basis로 진입시켜도 현재 목적값이 변하지 않는다는 뜻이다. 복수 최적해는 최적 face 위의 점들이 같은 값을 준다는 의미이지 모든 feasible point가 최적이라는 뜻이 아니다. 대안 최적해는 reduced cost가 0인 비기저변수를 entering 후보로 잡아 pivot해 찾는다.
- 근거: `DM_PDF01:p002:L003`, node_id=`n_DM_PDF01.multiple_optima`.
- 자주 하는 오답: “복수 최적”을 “아무 해나 가능”으로 해석한다.


### 문항 6

- 정답: ratio test는 entering column의 양수 계수가 있는 행에서만 가능하다. 모든 제약행 계수가 0 이하이면 `x3`를 늘려도 현재 기저변수가 음수가 되는 경계가 나타나지 않는다. 따라서 infeasible이 아니라 unbounded로 판정한다.
- 현실 해석: 목적값을 개선하는 활동을 제한할 자원, 수요상한, 용량상한 같은 제약이 빠졌을 가능성이 있다.
- 근거: `DM_PDF01:p003:L003`, node_id=`n_DM_PDF01.unbounded_solution`.
- 자주 하는 오답: ratio 후보가 없다는 사실을 infeasible로 해석한다.


### 문항 7

- 정답: RHS를 entering column 값으로 나누어 한계를 본다는 틀은 맞다. 그러나 음수 계수를 절댓값으로 바꾸면 안 된다. 최소 수정안은 “entering column의 양수 계수 행만 `RHS/a_ij`로 비교한다”이다. 음수 계수 행을 후보로 넣으면 실제로 제한하지 않는 제약을 leaving row로 선택해 feasibility를 깨뜨릴 수 있다.
- 근거: `DM_PDF01:p001:L003`, node_id=`n_DM_PDF01.minimum_ratio_test`.
- 자주 하는 오답: 모든 행을 ratio test에 넣는다.
