## 0. 학습 범위

핵심 질문은 네 가지다.

| 질문 | 확인할 것 |
|---|---|
| 무엇이 비선형인가? | 목적함수, 제약식, 또는 둘 다 |
| Solver 결과를 믿을 수 있는가? | local optimum 가능성, initial solution 민감성 |
| 전체최적해가 보장되는가? | 최대화는 오목 목적함수와 볼록 가능영역, 최소화는 볼록 목적함수와 볼록 가능영역 |
| KKT는 무엇을 보장하는가? | 일반적으로 지역 최적 후보의 필요조건, 추가 조건이 있으면 충분조건 |

## 3. 비선형계획 핵심 기준표

| 개념 | 기준 |
|---|---|
| Nonlinear programming | 목적함수 또는 제약식 중 하나 이상이 비선형이면 NLP다. |
| GRG Nonlinear | 매끄러운 비선형 문제에서 지역 최적 조건을 찾는 Solver 방식이다. |
| Local optimum | 주변 가능해보다 좋지만 전체 가능해 중 최고라고 보장되지는 않는다. |
| Global optimum | 모든 가능해와 비교해 최고 또는 최저인 해다. |
| Initial solution sensitivity | 시작점이 달라지면 Solver가 다른 local optimum에 도달할 수 있다. |
| Hessian | 다변수 함수의 2차미분 행렬로 convexity/concavity를 판정한다. |
| KKT | primal feasibility, stationarity, dual feasibility, complementary slackness를 확인한다. |
| 충분조건 | 최대화는 concave objective와 convex feasible region, 최소화는 convex objective와 convex feasible region이 핵심이다. |

### 문제 1. 스마트 온실 투입량 최적화 - GRG와 초기해 민감성

도메인:

한 스마트 온실이 영양액 투입량 `X`와 LED 조명시간 `Y`를 결정한다. 생산반응은 체감수익을 갖는 비선형 함수로 근사된다. `X`, `Y`는 연속변수다.

기본 반응모형:

\[
Q(X,Y)=35X^{0.5}Y^{0.4}
\]

제약식:

\[
2X+Y\le 60,\quad X\ge0,\quad Y\ge0
\]

추가 변형 모형:

\[
F(X,Y)=35X^{0.5}Y^{0.4}+14\sin(0.55X)\cos(0.35Y)
\]

초기해 후보:

\[
(1,1),\ (10,10),\ (20,5),\ (3,35)
\]

요구:

1. decision variables `X`, `Y`를 정의하라.
2. 기본 반응모형에서 무엇이 비선형인지 설명하라.
3. objective function과 constraints를 완성하라.
4. Excel Solver 관점에서 changing cells, target cell, constraint cells, solving method, initial values를 설명하라.
5. 기본 반응모형의 feasible region이 convex인지 판정하라.
6. 기본 반응모형에서 objective가 concave인지 Hessian 또는 Cobb-Douglas 조건으로 판정하고, 전체최적성이 보장되는 이유를 설명하라.
7. 추가 변형 모형 `F`가 왜 nonconvex landscape를 만들 수 있는지 설명하라.
8. 초기해 `(1,1)`, `(10,10)`, `(20,5)`, `(3,35)`에서 GRG를 시작한다고 할 때, 왜 multi-start 진단이 필요한지 설명하라.
9. contour plot과 feasible line을 그리는 코드를 작성하라. 단, problem notebook에서는 최적점 계산 결과를 출력하지 말 것.
10. 다음 오답을 진단하라: “Solver가 해를 하나 보여줬으므로 추가 변형 모형의 전체 최적해가 증명되었다.”
11. ML loss landscape 관점에서 기본 모형과 추가 변형 모형을 비교하라. convex/concave landscape와 nonconvex landscape의 차이를 설명하라.

```python
# 힌트 전용 시각화 scaffold: 문제 1 contour, gradient, normal, curvature
# 주의: 이 셀은 최적점을 계산하거나 출력하지 않는다.
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 30, 220)
y = np.linspace(0, 60, 260)
X, Y = np.meshgrid(x, y)
feasible = (2*X + Y <= 60)

Q = 35*np.sqrt(X)*np.power(Y, 0.4)
F = Q + 14*np.sin(0.55*X)*np.cos(0.35*Y)
Q_masked = np.where(feasible, Q, np.nan)
F_masked = np.where(feasible, F, np.nan)

starts = np.array([[1, 1], [10, 10], [20, 5], [3, 35]])
probe = np.array([8.0, 20.0])  # 예시점: 최적점이 아니라 curvature 해석용 점
normal = np.array([2.0, 1.0])  # 2X+Y<=60의 boundary normal
angles = np.linspace(0, 2*np.pi, 240)

def hessian_Q(xv, yv):
    fxx = 35*0.5*(-0.5)*(xv**(-1.5))*(yv**0.4)
    fyy = 35*0.4*(-0.6)*(xv**0.5)*(yv**(-1.6))
    fxy = 35*0.5*0.4*(xv**(-0.5))*(yv**(-0.6))
    return np.array([[fxx, fxy], [fxy, fyy]])

H_probe = hessian_Q(*probe)
dirs = np.c_[np.cos(angles), np.sin(angles)]
curv = np.einsum('bi,ij,bj->b', dirs, H_probe, dirs)

fig, axes = plt.subplots(2, 2, figsize=(12, 9), constrained_layout=True)
for ax, Z, title in [
    (axes[0, 0], Q_masked, '기본 Q contour: 단일 봉우리 구조를 의심해 보기'),
    (axes[0, 1], F_masked, '추가 F contour: 요동과 local optimum 가능성'),
]:
    cs = ax.contourf(X, Y, Z, levels=24, cmap='viridis')
    ax.plot(x, 60 - 2*x, color='white', linewidth=2, label='2X+Y=60')
    ax.scatter(starts[:, 0], starts[:, 1], color='red', s=35, label='초기해 후보')
    ax.set_xlim(0, 30)
    ax.set_ylim(0, 60)
    ax.set_xlabel('X: 영양액 투입량')
    ax.set_ylabel('Y: LED 조명시간')
    ax.set_title(title)
    ax.legend(loc='upper right')
    fig.colorbar(cs, ax=ax, shrink=0.82)

ax = axes[1, 0]
ax.fill([0, 30, 0], [0, 0, 60], color='#E8F4FF', alpha=0.7, label='feasible region')
ax.plot(x, 60 - 2*x, color='black', linewidth=2, label='binding boundary')
boundary_point = np.array([10.0, 40.0])
normal_scaled = normal / np.linalg.norm(normal) * 8
ax.arrow(boundary_point[0], boundary_point[1], normal_scaled[0], normal_scaled[1],
         head_width=1.2, length_includes_head=True, color='crimson', label='constraint normal (2,1)')
ax.scatter([boundary_point[0]], [boundary_point[1]], color='black', s=45)
ax.text(1, 52, 'KKT 힌트:\nmax에서 boundary가 binding이면\n∇Q와 λ(2,1)의 방향을 비교한다.', fontsize=10,
        bbox=dict(boxstyle='round,pad=0.35', facecolor='white', alpha=0.9))
ax.set_xlim(0, 30)
ax.set_ylim(0, 60)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('KKT 기하 scaffold: gradient와 constraint normal')
ax.legend(loc='lower left')

ax = axes[1, 1]
ax.plot(np.degrees(angles), curv, color='navy')
ax.axhline(0, color='black', linewidth=1)
ax.set_xlabel('방향 각도 theta')
ax.set_ylabel('v.T @ Hessian @ v')
ax.set_title('예시점에서 directional curvature: 모두 음수인지 확인')
ax.text(8, np.nanmin(curv)*0.85, '관찰/원인/제한/결론은 답안에 작성', fontsize=10,
        bbox=dict(boxstyle='round,pad=0.35', facecolor='white', alpha=0.9))
plt.show()
```

### 문제 2. 구독형 앱 가격결정 - 비선형 가격모형, Hessian, 전체최적성 검사

한 회사가 두 구독상품의 가격을 결정한다. 상품 1 가격을 `p1`, 상품 2 가격을 `p2`라고 한다. 수요는 가격에 따라 선형적으로 변하지만, 이익은 가격과 수요의 곱이므로 비선형이 된다.

수요함수:

\[
d_1=140-4p_1+1.5p_2
\]

\[
d_2=120+p_1-3p_2
\]

단위 원가:

\[
c_1=8,\quad c_2=10
\]

총이익:

\[
Profit=(p_1-8)d_1+(p_2-10)d_2
\]

제약식:

\[
d_1+d_2\le180
\]

\[
15\le p_1\le50,\quad 12\le p_2\le45
\]

\[
d_1\ge0,\quad d_2\ge0
\]

요구:

1. decision variables를 정의하라.
2. 왜 수요함수는 선형인데 총이익 함수는 비선형인지 설명하라.
3. objective function을 `p1, p2`만의 함수로 전개하라.
4. Solver 언어로 changing cells, target cell, constraint cells, solving method를 설명하라.
5. Hessian matrix를 계산하라.
6. Hessian이 negative definite인지 판정하라.
7. 최대화 문제에서 objective가 concave이면 어떤 의미인지 설명하라.
8. 제약식들이 모두 선형일 때 feasible region이 convex인 이유를 설명하라.
9. 이 문제에서 KKT 조건이 전체최적성까지 보장되는지 판정하라.
10. GRG 결과와 Hessian/global optimality check를 어떻게 함께 써야 하는지 설명하라.
11. 가격결정 문제를 ML 관점에서 “예측된 반응식 위에서 의사결정변수 `p`를 최적화하는 문제”로 해석하라. 단, 실제 ML 학습 문제로 바꾸지 말 것.

```python
# 힌트 전용 시각화 scaffold: 문제 2 profit contour, Hessian curvature, slack 읽기
# 주의: 이 셀은 최적 가격을 계산하거나 출력하지 않는다.
import numpy as np
import matplotlib.pyplot as plt

p1 = np.linspace(15, 50, 260)
p2 = np.linspace(12, 45, 260)
P1, P2 = np.meshgrid(p1, p2)
D1 = 140 - 4*P1 + 1.5*P2
D2 = 120 + P1 - 3*P2
Profit = (P1 - 8)*D1 + (P2 - 10)*D2
feasible = (D1 + D2 <= 180) & (D1 >= 0) & (D2 >= 0)
Profit_masked = np.where(feasible, Profit, np.nan)

H = np.array([[-8.0, 2.5], [2.5, -6.0]])
angles = np.linspace(0, 2*np.pi, 240)
dirs = np.c_[np.cos(angles), np.sin(angles)]
curv = np.einsum('bi,ij,bj->b', dirs, H, dirs)

example = np.array([25.0, 25.0])  # 예시 가격점: 최적점이 아니라 constraint 읽기 연습용
ex_d1 = 140 - 4*example[0] + 1.5*example[1]
ex_d2 = 120 + example[0] - 3*example[1]
slacks = {
    '총수요': 180 - (ex_d1 + ex_d2),
    'p1 하한': example[0] - 15,
    'p1 상한': 50 - example[0],
    'p2 하한': example[1] - 12,
    'p2 상한': 45 - example[1],
    'd1>=0': ex_d1,
    'd2>=0': ex_d2,
}

fig, axes = plt.subplots(1, 3, figsize=(15, 4.8), constrained_layout=True)
ax = axes[0]
cs = ax.contourf(P1, P2, Profit_masked, levels=28, cmap='magma')
ax.contour(P1, P2, feasible.astype(int), levels=[0.5], colors='white', linewidths=2)
ax.scatter([example[0]], [example[1]], color='cyan', edgecolor='black', s=55, label='slack 예시점')
ax.set_xlabel('p1')
ax.set_ylabel('p2')
ax.set_title('Profit contour와 feasible boundary')
ax.legend(loc='upper right')
fig.colorbar(cs, ax=ax, label='Profit')

ax = axes[1]
ax.plot(np.degrees(angles), curv, color='darkgreen')
ax.axhline(0, color='black', linewidth=1)
ax.set_xlabel('방향 각도 theta')
ax.set_ylabel('v.T @ H @ v')
ax.set_title('Hessian directional curvature: 음수이면 concave')

ax = axes[2]
labels = list(slacks.keys())
values = list(slacks.values())
ax.barh(labels, values, color='#6BAED6')
ax.axvline(0, color='black', linewidth=1)
ax.set_title('예시점 constraint slack 읽기')
ax.set_xlabel('slack')
ax.text(0.02, 0.04, 'KKT 힌트: 최적점에서\nslack>0이면 해당 multiplier=0', transform=ax.transAxes,
        bbox=dict(boxstyle='round,pad=0.35', facecolor='white', alpha=0.9))
plt.show()
```

```python
# 힌트 전용 시각화 scaffold: 문제 3 KKT normal, contour, multi-start 후보
# 주의: 이 셀은 최적 예산이나 multiplier 값을 계산하거나 출력하지 않는다.
import numpy as np
import matplotlib.pyplot as plt

x1 = np.linspace(0, 45, 260)
x2 = np.linspace(0, 45, 260)
X1, X2 = np.meshgrid(x1, x2)
feasible = (X1 + X2 <= 45)
R = 90*np.log1p(X1) + 75*np.log1p(X2) - 1.8*X1 - 1.2*X2
R_alt = R + 8*np.sin(0.35*X1)*np.cos(0.30*X2)
starts = np.array([[0, 0], [5, 5], [15, 10], [10, 30], [30, 10], [22, 22]])

fig, axes = plt.subplots(1, 3, figsize=(15.5, 4.8), constrained_layout=True)
for ax, Z, title in zip(axes[:2], [R, R_alt], ['기본 R contour', '대안 R_alt contour']):
    Z_masked = np.where(feasible, Z, np.nan)
    cs = ax.contourf(X1, X2, Z_masked, levels=28, cmap='cividis')
    ax.plot(x1, 45 - x1, color='white', linewidth=2, label='x1+x2=45')
    ax.scatter(starts[:, 0], starts[:, 1], color='red', s=32, label='multi-start 후보')
    ax.set_xlim(0, 45)
    ax.set_ylim(0, 45)
    ax.set_xlabel('x1: 채널 1 예산')
    ax.set_ylabel('x2: 채널 2 예산')
    ax.set_title(title)
    ax.legend(loc='upper right')
    fig.colorbar(cs, ax=ax, shrink=0.85)

ax = axes[2]
ax.fill([0, 45, 0], [0, 0, 45], color='#F1F7E8', alpha=0.85, label='feasible region')
ax.plot(x1, 45-x1, color='black', linewidth=2, label='binding budget line')
normal = np.array([1.0, 1.0])
normal = normal / np.linalg.norm(normal) * 7
base = np.array([18.0, 27.0])
ax.arrow(base[0], base[1], normal[0], normal[1], color='crimson', head_width=1.0,
         length_includes_head=True, label='budget normal (1,1)')
ax.text(2, 35, 'KKT 힌트:\n예산 제약이 binding이면\n∇R = λ(1,1)를 확인한다.\nλ는 예산 RHS의 한계가치다.', fontsize=10,
        bbox=dict(boxstyle='round,pad=0.35', facecolor='white', alpha=0.92))
ax.set_xlim(0, 45)
ax.set_ylim(0, 45)
ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_title('라그랑주승수 기하 scaffold')
ax.legend(loc='lower left')
plt.show()
```
