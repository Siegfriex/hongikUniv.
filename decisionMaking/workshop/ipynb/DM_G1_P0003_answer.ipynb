## 2.5 시각화 심화 학습 보드

이 섹션은 P0003의 Big-M과 duality bridge를 시각적으로 정리한다. 핵심은 다음이다.

1. **Big-M은 artificial variable을 벌점으로 밀어내는 장치다.**
2. **dual problem은 primal의 제약과 변수를 서로 바꿔 보는 가격 해석이다.**
3. **complementary slackness는 최적해에서 slack과 가격이 동시에 양수일 수 없다는 조건이다.**


```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch, Polygon

# Matplotlib에 확실히 존재하는 DejaVu Sans를 사용한다.
# 그림 내부 텍스트는 폰트 깨짐 방지를 위해 ASCII/English로 제한하고,
# 한국어 해설은 Markdown 셀에서 제공한다.
mpl.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.unicode_minus": False,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "savefig.facecolor": "white",
    "axes.titleweight": "bold",
    "axes.titlesize": 13,
    "axes.labelsize": 10,
})

def box(ax, xy, wh, text, fc="#eef6ff", ec="#243238", fs=9):
    x, y = xy
    w, h = wh
    ax.add_patch(Rectangle((x, y), w, h, facecolor=fc, edgecolor=ec, lw=1.4))
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", fontsize=fs)

def arrow(ax, start, end, color="#2563eb"):
    ax.add_patch(FancyArrowPatch(start, end, arrowstyle="-|>", mutation_scale=14, lw=1.6, color=color))

def draw_flow(ax, rows, title):
    ax.axis("off")
    ax.set_title(title)
    n = len(rows)
    colors = ["#e7f0fd", "#e9f7ef", "#fff4d6", "#fce8e6", "#ece7fb", "#e8f6f8"]
    centers = []
    for i, row in enumerate(rows):
        y = 0.90 - i * (0.78 / max(n-1, 1))
        row_centers = []
        m = len(row)
        for j, txt in enumerate(row):
            x = 0.08 + j * (0.84 / max(m-1, 1)) if m > 1 else 0.38
            box(ax, (x, y-0.055), (0.24, 0.11), txt, fc=colors[(i+j) % len(colors)], fs=8.5)
            row_centers.append((x+0.12, y))
        centers.append(row_centers)
    for i in range(len(centers)-1):
        for a in centers[i]:
            for b in centers[i+1]:
                if len(centers[i]) == 1 or len(centers[i+1]) == 1 or abs(a[0]-b[0]) < 0.28:
                    arrow(ax, (a[0], a[1]-0.06), (b[0], b[1]+0.06))

print("Visualization setup ready. Matplotlib font:", mpl.rcParams["font.family"][0])
```

```python
# P0003 Board 1: Big-M sign and artificial-zero condition
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

ax = axes[0]
r = np.linspace(0, 4, 100)
ax.plot(r, -r, label="penalty: -M*r for max form", lw=3)
ax.plot(r, r, label="wrong reward: +M*r", lw=3, ls="--")
ax.axhline(0, color="#111", lw=1)
ax.set_xlabel("artificial variable r")
ax.set_ylabel("objective effect / M")
ax.set_title("Big-M sign: artificial must hurt objective")
ax.grid(alpha=0.25); ax.legend()

ax = axes[1]
names = ["r1 final", "r2 final", "allowed"]
vals = [0, 3, 0]
ax.bar(names, vals, color=["#60a5fa", "#f87171", "#93c5fd"])
ax.text(1, 2.0, "positive artificial\n=> not original feasible", ha="center", fontsize=9)
ax.set_title("Artificial variables must return to zero")
ax.set_ylabel("artificial value")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# P0003 Board 2: primal-dual mapping and transposed coefficient structure
fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

ax = axes[0]
A = np.array([[8, 4], [4, 4], [4, 0], [0, 1]], dtype=float)
im = ax.imshow(A, cmap="YlGnBu")
ax.set_xticks([0,1]); ax.set_xticklabels(["x1", "x2"])
ax.set_yticks(range(4)); ax.set_yticklabels(["constraint1", "constraint2", "constraint3", "constraint4"])
for i in range(A.shape[0]):
    for j in range(A.shape[1]):
        ax.text(j, i, f"{A[i,j]:g}", ha="center", va="center")
ax.set_title("Primal matrix A: constraints x variables")
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

draw_flow(
    axes[1],
    [
        ["Primal constraints"],
        ["Dual variables y"],
        ["Primal variables x"],
        ["Dual constraints"],
        ["RHS -> dual objective"],
        ["Objective coefficients -> dual RHS"]
    ],
    "Primal-dual correspondence"
)
plt.tight_layout()
plt.show()
```

```python
# P0003 Board 3: duality gap, complementary slackness, and sensitivity routing
fig, axes = plt.subplots(1, 3, figsize=(18, 5.2))

ax = axes[0]
ax.bar(["Primal Z", "Dual W"], [700, 1540], color=["#fb923c", "#60a5fa"])
ax.plot([0, 1], [700, 1540], color="#7c3aed", lw=2)
ax.text(0.5, 1120, "duality gap", ha="center", fontsize=9)
ax.set_title("Weak duality: Z <= W for max/min pair")
ax.set_ylabel("objective value")
ax.grid(axis="y", alpha=0.25)

ax = axes[1]
items = ["slack", "dual y", "product"]
vals = [40, 3, 120]
ax.bar(items, vals, color=["#93c5fd", "#facc15", "#f87171"])
ax.text(2, 80, "must be 0\nat optimum", ha="center", fontsize=9)
ax.set_title("Complementary slackness check")
ax.grid(axis="y", alpha=0.25)

draw_flow(
    axes[2],
    [
        ["User question"],
        ["Variable/activity?"],
        ["Reduced cost"],
        ["RHS/resource?"],
        ["Shadow price"],
        ["Check allowable range"]
    ],
    "Sensitivity report routing"
)
plt.tight_layout()
plt.show()
```
