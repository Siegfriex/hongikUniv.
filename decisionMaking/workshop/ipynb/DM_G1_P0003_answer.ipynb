- 근거: `DM_PDF05:p004:L002`, `DM_PDF01:p005:L001`.


### 문항 4

- 정답: 맞는 부분은 M이 artificial을 basis에서 빼내기 위한 장치라는 점이다. 그러나 max 형태에서 `+M r`가 artificial을 보상하는 방식이면 위험하다. Big-M penalty의 목적은 artificial이 남는 해를 매우 불리하게 만드는 것이다. 안전한 절차는 목적이 max인지 min인지 확인하고, artificial이 커질수록 objective가 나빠지는 방향으로 부호를 정하는 것이다.
