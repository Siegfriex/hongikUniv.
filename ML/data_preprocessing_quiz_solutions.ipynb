빈칸을 채운 완성본입니다.

빈칸을 채워 함수를 완성하고 실행하시오.
- 모든 연산은 이미지 전체 픽셀 기준으로 수행합니다.

```python
import numpy as np

img = np.array([
    [  0,  50, 100, 150, 200],
    [ 25,  75, 125, 175, 225],
    [ 50, 100, 150, 200, 250],
    [ 75, 125, 175, 225, 255],
    [100, 150, 200, 250, 255]
], dtype=float)

print(img)
```

## (1) Normalize 함수 완성
힌트: min, max 사용

**풀이:** 최소–최대 정규화로 \([0,1]\) 구간으로 선형 변환.  
`(x - x.min()) / (x.max() - x.min())`

```python
def normalize(x):
    return (x - x.min()) / (x.max() - x.min())

norm_img = normalize(img)
print(norm_img)
```

## (2) Standardize 함수 완성
힌트: mean, std 사용

**풀이:** 평균 0, 표준편차 1에 가깝게 맞추는 z-score.  
`(x - x.mean()) / x.std()`  
(NumPy `std` 기본은 전체 모집단 분산에 대한 `ddof=0`.)

```python
def standardize(x):
    return (x - x.mean()) / x.std()

std_img = standardize(img)
print(std_img.mean())
```

## (3) 결측값 처리
힌트: np.nanmean, np.isnan 사용

**풀이:** 결측을 제외한 평균으로 채움.  
- `np.nanmean(x)`: NaN을 빼고 평균  
- `np.where(np.isnan(x), mean_val, x)`: NaN이면 평균, 아니면 원값

```python
img2 = img.copy()
img2[0, 0] = np.nan

def fill_missing(x):
    mean_val = np.nanmean(x)
    return np.where(np.isnan(x), mean_val, x)

filled = fill_missing(img2)
print(filled)
```

## (4) 개념 문제
왜 정규화를 사용하는가?

**예시 답안:**
- **스케일 통일**: 픽셀·특성마다 값의 범위가 크게 다르면, 경사하강 등 최적화가 한 축에 치우치거나 수렴이 느려질 수 있다. 정규화(또는 표준화)로 비슷한 척도에 맞춘다.
- **수치 안정성**: 큰 값이 그래디언트·활성함수 입력을 불안정하게 만들 수 있어, 학습·추론을 안정시키는 데 도움이 된다.
- **비교·가시화**: 서로 다른 이미지·채널을 같은 규칙으로 맞추면 비교와 시각화가 쉬워진다.
