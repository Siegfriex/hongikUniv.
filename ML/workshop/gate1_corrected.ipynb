**target** =

- `1 = 매우불만족`
- `2 = 불만족`
- `3 = 보통`
- `4 = 만족`
- `5 = 매우만족`

**이 숫자는 크기/거리 의미가 있는가?**  
순서 의미는 있다. `1 < 2 < 3 < 4 < 5`이므로 5점이 1점보다 만족도가 높다는 해석은 가능하다.  
하지만 `1 -> 2`의 차이와 `4 -> 5`의 차이가 정확히 같은 심리적 거리라고 단정하기는 어렵다.

**class 이름을 숫자로 바꾼 것인가?**

**문제 유형**  
가장 자연스러운 해석은 순서형 분류, 즉 ordinal classification이다.

**다른 방식으로 풀 수 있는가?**  
가능하다.

1. 다중분류: `1,2,3,4,5`를 다섯 class로 본다.
2. 순서형 분류: `1 < 2 < 3 < 4 < 5`의 순서를 반영한다.
3. 회귀: 만족도를 거의 연속 점수처럼 보고 예측한 뒤 반올림/clip 처리한다.

> **이전 오답 주석**
>
> - `X =`, `y =`를 비워 둔 것은 답안칸 기준 누락이다.
> - “이 숫자는 크기/거리 의미가 있는가?”에서 feature들의 단위 차이를 설명한 것이 핵심 오류였다. 여기서 판단해야 하는 숫자는 `y = 1~5` 만족도 점수다.
> - “5중 분류”는 가능하지만 최선 답안은 아니다. 이 문제의 핵심 표현은 `순서형 target`, `ordinal classification`이다.


## C-3. 영화 장르 encoding

### 문제

영화 장르를 다음처럼 저장했다.

- 0 = 액션
- 1 = 로맨스
- 2 = 스릴러
- 3 = 다큐멘터리

줄거리 텍스트를 이용해 장르 번호를 예측한다.

### 답할 것

- X =
- y =
- feature =
- target =
- 이 숫자는 크기/거리 의미가 있는가?
- class 이름을 숫자로 바꾼 것인가?
- 문제 유형 =
- 회귀로 풀면 생기는 문제 =

```python
import pandas as pd
import numpy as np
import re

target_names = np.array(["액션", "로맨스", "스릴러", "다큐멘터리"])
raw_df = pd.DataFrame({
    "plot": [
        "특수부대 요원이 폭발과 총격 액션 속에서 테러리스트를 추격한다.",
        "두 사람이 우연히 만나 사랑과 로맨스를 키우지만 이별 위기에 놓인다.",
        "연쇄 살인범을 쫓는 형사가 어두운 도시에서 스릴러 같은 긴장을 겪는다.",
        "북극곰과 기후 변화를 담은 다큐멘터리 영화가 자연의 아름다움을 보여준다.",
    ],
    "genre_code": [0, 1, 2, 3],
})
keywords = ["액션", "로맨스", "스릴러", "다큐멘터리"]

def extract_keyword_features(text, keywords):
     words = re.findall(r"[가-힣]+", text)
     total_words = max(len(words), 1)
     result = {}
     for kw in keywords:
        count = text.count(kw)
        result[f"{kw}_exists"] = int(count > 0)
        result[f"{kw}_count"] = count
        result[f"{kw}_ratio"] = count / total_words
     return result

feature_df = raw_df["plot"].apply(lambda t: pd.Series(extract_keyword_features(t, keywords)))

X_df = feature_df[[f"{kw}_ratio" for kw in keywords]].copy()
X_df.columns = [f"feature_{i+1}_{kw}" for i, kw in enumerate(keywords)]
X = X_df.values
y = raw_df["genre_code"].values
y_s = pd.Series(y, name="genre_code")
print(X_df)
print(feature_df)
print("target 이름:", target_names)
print("y 고유값:", np.unique(y))
print("클래스별 개수:")
print(y_s.value_counts().sort_index())

print("=== 로데이터 (줄거리 + 장르 code) ===")
print(raw_df.head())



```

### 정답 C-3: 영화 장르 encoding

이 정답에서는 최종 입력 `X`를 코드 출력에 맞춰 장르 키워드 ratio 4개로 고정한다.

**X** = 영화 설명 텍스트에서 만든 장르 키워드 ratio feature 행렬

**feature** =

- `feature_1_액션`
- `feature_2_로맨스`
- `feature_3_스릴러`
- `feature_4_다큐멘터리`

**y** = 영화 장르 class code  
예: `0, 1, 2, 3`

**target** =

- `0 = 액션`
- `1 = 로맨스`
- `2 = 스릴러`
- `3 = 다큐멘터리`

**이 숫자는 크기/거리 의미가 있는가?**  
아니다. `0, 1, 2, 3`은 장르 사이의 크기나 거리를 뜻하지 않는다. 액션이 로맨스보다 작거나, 다큐멘터리가 스릴러보다 크다는 의미가 아니다.

**class 이름을 숫자로 바꾼 것인가?**  
그렇다. `액션/로맨스/스릴러/다큐멘터리`라는 class 이름을 `0,1,2,3`으로 인코딩한 것이다.

**문제 유형**  
명목형 다중분류 문제다.

**회귀로 풀면 생기는 문제**  
회귀는 장르 코드 사이에 순서와 거리 의미가 있다고 가정한다. 예측값이 `1.6`처럼 나오면 어떤 장르인지 애매하다. 또한 MSE는 `액션(0) -> 다큐멘터리(3)` 오류를 `액션(0) -> 로맨스(1)` 오류보다 더 큰 수치 오차로 처리하지만, 장르 class에는 그런 거리 의미가 없다.

> **이전 오답 주석**
>
> - `exists/count/ratio` feature engineering 자체는 좋은 시도였지만, 답안에서 최종 X가 ratio 4개인지 12개인지 불일치했다. 정답에서는 ratio 4개로 고정했다.
> - “이 숫자”를 feature의 `exists/count/ratio` 숫자로 해석한 것이 핵심 오류였다. 여기서 판단해야 하는 숫자는 `y = 0,1,2,3` 장르 코드다.
> - `class 이름을 숫자로 바꾼 것인가?`를 비워 둔 것은 누락이다. 정답은 “그렇다”다.
> - “결정계수”라는 표현은 부적절하다. 결정계수 `R^2`는 회귀 평가 지표이지 class를 정하는 기준이 아니다.

```python
import pandas as pd
import numpy as np

target_names = ["원활", "보통", "혼잡", "매우혼잡"]

#피처1 : 차량수 / 도로용량 비율
#피처2 : 시간대별 가중치/요일별 가중치 
#피처3 : 평균속도/도로구간 비율
#피처4 : 날씨별 가중치 
#피처5 : 사고여부/도로구간 비율
#피처6 : 차량수 / 시간대별 기대용량 비율

## 피처 정의표 (Gate 답안용)
feature_spec = pd.DataFrame({
    "피처": ["feature_1", "feature_2", "feature_3", "feature_4", "feature_5", "feature_6"],
    "한글명": [
        "차량수/도로용량 비율",
        "시간대가중치/요일가중치",
        "평균속도/도로길이 비율",
        "날씨 가중치",
        "사고여부/도로길이 비율",
        "차량수/시간대용량 비율",
    ],
    "수식": [
        "vehicle_count / road_capacity",
        "time_weight[hour] / day_weight[dayofweek]",
        "avg_speed / road_km",
        "weather_weight[weather]",
        "accident / road_km",
        "vehicle_count / hour_capacity[hour]",
    ],
    "타입": ["비율", "비율", "비율", "가중치", "비율", "비율"],
})


# 보조 매핑 (로데이터를 사람이 읽게)
day_names = ["월", "화", "수", "목", "금", "토", "일"]
hour_bucket = {
    range(0, 6): "새벽",
    range(6, 10): "출근",
    range(10, 17): "낮",
    range(17, 20): "퇴근",
    range(20, 24): "야간",
}
def get_hour_bucket(h):
    for r, name in hour_bucket.items():
        if h in r:
            return name
    return "기타"

# 1) 로데이터 직접 제작
raw_df = pd.DataFrame({
    "vehicle_count": [80, 150, 320, 410, 200, 90, 380, 420],
    "hour":          [3, 10, 8, 18, 14, 6, 19, 8],
    "dayofweek":     [1, 2, 1, 4, 3, 6, 5, 1],
    "weather":       ["맑음", "맑음", "비", "흐림", "맑음", "맑음", "비", "눈"],
    "avg_speed":     [70, 55, 25, 15, 45, 65, 20, 10],
    "road_km":       [3.0, 5.0, 4.0, 6.0, 5.0, 3.0, 4.0, 5.0],
    "road_capacity": [200, 300, 400, 400, 300, 200, 400, 400],
    "accident":      [0, 0, 1, 1, 0, 0, 1, 1],
    "congestion_code": [0, 1, 2, 3, 1, 0, 2, 3],
})


time_weight = {h: 1.0 for h in range(24)}
for h, w in {7:1.5, 8:2.0, 9:1.8, 17:1.9, 18:2.2, 19:1.7}.items():
    time_weight[h] = w

day_weight = {0:1.0, 1:1.0, 2:1.0, 3:1.0, 4:1.0, 5:0.7, 6:0.6}

weather_weight = {"맑음":1.0, "흐림":1.1, "비":1.3, "눈":1.5}

hour_capacity = {h: (300 if h in [7,8,17,18,19] else 200) for h in range(24)}

def extract_traffic_features(row):
    return pd.Series({
        "feature_1_vehicle_capacity_ratio": row["vehicle_count"] / row["road_capacity"],
        "feature_2_time_day_weight_ratio": time_weight[row["hour"]] / day_weight[row["dayofweek"]],
        "feature_3_speed_road_ratio": row["avg_speed"] / row["road_km"],
        "feature_4_weather_weight": weather_weight[row["weather"]],
        "feature_5_accident_road_ratio": row["accident"] / row["road_km"],
        "feature_6_vehicle_hour_ratio": row["vehicle_count"] / hour_capacity[row["hour"]],
    })


# 2) 피처 추출
feature_df = raw_df.apply(extract_traffic_features, axis=1)
X_df = feature_df.copy()
X = X_df.values
y = raw_df["congestion_code"].values
y_s = pd.Series(y, name="congestion_code")

# 통합 테이블: 로데이터 + 해석 컬럼 + 피처 + 정답
df = raw_df.copy()
df["요일"] = df["dayofweek"].map(lambda d: day_names[d])
df["시간대구간"] = df["hour"].map(get_hour_bucket)
df["혼잡도"] = df["congestion_code"].map(dict(enumerate(target_names)))
df = pd.concat([df, X_df], axis=1)

# 클래스 코드 → 한글 라벨 매핑 (rename(index=배열)은 Series에서 오류)
class_counts = y_s.value_counts().sort_index()
class_counts.index = [target_names[i] for i in class_counts.index]

print("=" * 60)
print("【1】 피처 정의")
print(feature_spec.to_string(index=False))

print("\n" + "=" * 60)
print("【2】 로데이터 + 분류 라벨")
print(df[[
    "vehicle_count", "hour", "시간대구간", "dayofweek", "요일",
    "weather", "avg_speed", "road_km", "accident",
    "congestion_code", "혼잡도"
]].to_string(index=True))

print("\n" + "=" * 60)
print("【3】 X / y 크기")
print(f"X shape: {X.shape}  → {X.shape[0]}개 샘플, {X.shape[1]}개 피처")
print(f"y shape: {y.shape}  → 클래스 코드 {list(np.unique(y))}")
print(f"target 이름: {list(target_names)}")

print("\n" + "=" * 60)
print("【4】 클래스별 개수")
print(class_counts)

print("\n" + "=" * 60)
print("【5】 피처 통계")
print(X_df.describe().round(3).to_string())

print("\n" + "=" * 60)
print("【6】 샘플별 한 줄 요약")
for i, row in df.iterrows():
    print(
        f"[{i}] {row['혼잡도']}({row['congestion_code']}) | "
        f"{row['hour']}시({row['시간대구간']}) {row['요일']} | "
        f"차량 {row['vehicle_count']}대 | 속도 {row['avg_speed']}km/h | "
        f"날씨 {row['weather']} | "
        f"f1={row['feature_1_vehicle_capacity_ratio']:.3f}, "
        f"f6={row['feature_6_vehicle_hour_ratio']:.3f}"
    )

print("\n" + "=" * 60)
print("【7】 Gate 판단 요약")
print("X =", list(X_df.columns))
print("y = congestion_code (0~3)")
print("feature = 위 6개 수치 피처 (비율/가중치)")
print("target =", list(target_names), "← class 이름을 숫자로 인코딩")
print("문제 유형 = 순서형 다중분류 (원활 < 보통 < 혼잡 < 매우혼잡)")
print("회귀로 풀면 = 2.3 같은 중간값 예측, 클래스 경계 모호")

df
```

## 보완 1. 성별 0=남성, 1=여성

### 정답

**X** = 성별을 예측하기 위한 입력 feature들  
예: 키, 체중, 음성 특징, 설문 응답 등 문제에서 주어진 예측변수

**y** = 성별 class

**target** =

- `0 = 남성`
- `1 = 여성`

**숫자 의미**  
`0`과 `1`은 크기나 거리 의미가 없다. 남성이 여성보다 작거나, 여성이 남성보다 1만큼 큰 것이 아니다.

**class encoding 여부**  
성별이라는 명목형 class를 `0/1`로 인코딩한 것이다.

**문제 유형**  
명목형 이진분류 문제다. 표현은 `binary classification` 또는 `Bernoulli target`이 적절하다.

**회귀로 풀면 생기는 문제**  
회귀로 풀면 `0.4`, `0.7` 같은 연속 출력이 나올 수 있다. 이 값은 “성별의 중간값”이라는 실제 수치 의미를 갖지 않는다. 분류 모델에서는 이런 값을 class 1일 확률로 해석하고 threshold로 최종 class를 정한다.

### 이전 오답 주석

- 전체 방향은 거의 맞았다.
- `바이노미널`보다는 `이진분류(binary classification)` 또는 `Bernoulli target`이 더 안전한 표현이다.
- 로지스틱 회귀나 sigmoid classifier의 `0.7`은 성별의 회귀값이 아니라 class 1의 확률이라는 점을 더 명확히 해야 한다.

### 최종 답안형

`0/1`은 성별 class를 숫자로 인코딩한 것이며 크기나 거리 의미가 없다. 따라서 이 문제는 명목형 이진분류다. 회귀로 풀면 `0.4`, `0.7` 같은 연속 출력이 나오며, 이는 성별이라는 class의 실제 수치값으로 해석될 수 없다. 분류 모델에서는 이 값을 class 1의 확률로 해석하고 threshold로 최종 class를 정한다.


## 보완 2. 안전등급 0=정상, 1=주의, 2=위험, 3=매우위험

### 정답

**X** = 안전등급을 예측하기 위한 입력 feature들  
예: 센서값, 이상 징후 수, 온도, 진동, 압력, 사고 이력 등

**y** = 안전등급 class

**target** =

- `0 = 정상`
- `1 = 주의`
- `2 = 위험`
- `3 = 매우위험`

**숫자 의미**  
순서 의미는 있다. `정상 < 주의 < 위험 < 매우위험`이라는 위험도 방향이 존재한다.

하지만 거리/간격 의미는 불명확하다. `정상 -> 주의`의 차이와 `위험 -> 매우위험`의 차이가 실제 위험도에서 정확히 같은 간격이라고 보장할 수 없다.

**class encoding 여부**  
class 이름을 숫자로 인코딩한 것이다. 다만 순서가 있는 class encoding이다.

**문제 유형**  
순서형 분류, 즉 ordinal classification이 가장 자연스럽다.

**회귀로 풀면 생기는 문제**  
일반 회귀로 풀면 `0,1,2,3` 사이의 간격이 실제 위험도 차이와 동일하다고 가정하게 된다. 이 가정은 강하다. 대신 정상에서 매우위험으로 틀리는 오류를 더 크게 벌주는 ordinal loss나 cost-sensitive classification은 가능하다.

### 이전 오답 주석

- `0 < 1 < 2 < 3`의 순서 의미와 간격 불명확성을 잡은 것은 좋았다.
- 다만 “분류된 것에 대해 가중을 매길 수 있다”는 말은 모델링 설계 선택으로 정리해야 한다.
- 가중/비용을 줄 수 있다는 사실이 곧 `0,1,2,3`을 단순 회귀 target처럼 써도 된다는 뜻은 아니다.

### 최종 답안형

안전등급 `0/1/2/3`은 class encoding이다. 다만 `정상 < 주의 < 위험 < 매우위험`이라는 순서가 있으므로 순서형 분류로 보는 것이 적절하다. 숫자 간격이 실제 위험도 차이와 동일하다고 보장되지 않으므로 일반 회귀로 풀면 거리 의미를 과하게 가정하게 된다. 다만 심각한 오분류에 더 큰 비용을 주는 ordinal loss나 cost-sensitive classification은 가능하다.


## 보완 3. 수학점수 0~100

### 정답

**X** = 수학점수를 예측하기 위한 입력 feature들  
예: 공부 시간, 출석률, 과제 점수, 이전 시험 점수, 문제 풀이량 등

**y** = 수학점수 `0~100`

**숫자 의미**  
실제 수치량이다. `80점`과 `90점`의 차이는 10점이고, `40점`과 `50점`의 차이도 10점이다.

**class encoding 여부**  
아니다. `0~100`은 class 이름을 숫자로 바꾼 것이 아니라 실제 측정된 점수값이다.

**문제 유형**  
기본적으로 회귀 문제다.

**등급화하면 어떻게 되는가?**

원래 점수 `0~100` 자체는 ordinal classification이 아니다. 회귀 target이다.

하지만 점수를 구간화하면 문제 유형이 바뀐다.

- 원점수 `0~100` 예측 = 회귀
- `90점 이상 A`, `80점 이상 B` 등으로 변환 = 순서형 다중분류
- `1~9등급`으로 변환 = 순서형 분류
- `합격/불합격`으로 변환 = 이진분류

등급으로 바꾼 순간 모델의 target은 더 이상 정확한 점수값이 아니라 구간 class다. 이 경우 class 사이에 순서는 있지만, 등급 코드 간 간격이 원래 점수 차이를 완전히 보존한다고 보기는 어렵다.

### 이전 오답 주석

- `수학점수 0~100 = 실제 수치량 = 회귀`까지는 맞았다.
- “순서형 인코딩으로서 2와 달리 거리가 의미가 있다”는 표현은 교정이 필요하다.
- 원점수는 거리 의미가 있지만, 점수를 등급으로 바꾼 순간 target은 구간 class가 된다.
- 특히 9등급제가 percentile 기반이면 등급 간 점수 폭이 동일하지 않을 수 있다.

### 최종 답안형

수학점수 `0~100`은 실제 수치량이므로 기본적으로 회귀 문제다. 분류로 바꾸려면 점수 구간을 미리 정의해야 한다. 예를 들어 `90점 이상 A`, `80점 이상 B`처럼 바꾸면 `y`는 순서형 class가 된다. 이 경우 class 사이에 순서는 있지만, 등급 코드 간 간격이 원래 점수 차이를 완전히 보존한다고 보기는 어렵다. 따라서 원점수를 예측하려면 회귀, 등급을 예측하려면 순서형 분류가 적절하다.


## 보완 6. 기온 예측

### 정답

**X** = 기온을 예측하기 위한 입력 feature들  
예: 날짜, 시간, 지역, 습도, 풍속, 기압, 계절, 과거 기온 등

**y** = 기온 값  
예: 섭씨 `-5°C`, `0°C`, `23.5°C`, `31°C`

**숫자 의미**  
기온은 실제 수치값이다. 음수도 가능하며, 값의 차이는 온도 차이를 의미한다.

**class encoding 여부**  
아니다. 기온은 class 이름을 숫자로 바꾼 것이 아니라 연속 수치형 target이다.

**문제 유형**  
기본적으로 회귀 문제다.

**척도 해석**

섭씨온도는 보통 등간척도에 가깝다. 섭씨 `0°C`는 물이 어는 기준점이지 “온도가 없음”을 뜻하지 않는다. 따라서 섭씨는 음수도 가능하다.

- 섭씨온도 = 등간척도에 가까움
- 켈빈온도 = 절대 0이 있으므로 비율척도에 가까움

**MSE와 음수 target**

MSE는 음수 target에도 그대로 사용할 수 있다. 수식을 수정할 필요가 없다.

예를 들어 실제 기온이 `-5°C`, 예측 기온이 `-3°C`라면 오차는 `-2`이고 제곱오차는 `4`다. 정상적으로 계산된다.

**출력층 주의**

수정해야 할 것은 loss가 아니라 output activation이다. 기온은 음수 예측이 가능해야 하므로 회귀 모델의 마지막 출력층은 보통 linear output, 즉 activation 없음이 적절하다. sigmoid나 ReLU처럼 출력 범위를 제한하는 activation은 부적절할 수 있다.

**분류로 바꾸는 경우**

기온을 `매우추움/추움/보통/더움/매우더움`처럼 구간화하면 순서형 분류 문제가 된다. 하지만 이때는 원래의 정확한 기온값을 예측하는 것이 아니라 온도 구간 class를 예측하는 것이다.

### 이전 오답 주석

- `기온은 실제 수치값`, `음수 가능`, `기본적으로 회귀`, `분류로 바꾸려면 급간화 가능`은 맞았다.
- `섭씨온도는 절대영도 기준`이라는 설명은 틀렸다. 섭씨 `0°C`는 절대적 0이 아니다.
- `수치형 범주`라는 표현은 부적절하다. 기온은 범주형이 아니라 연속 수치형 변수다.
- `비율척도 등간척도가 아니다`가 아니라, 섭씨온도는 등간척도에 가깝고 켈빈온도는 비율척도에 가깝다.
- 음수 target 때문에 MSE 수식을 수정해야 하는 것이 아니다. MSE는 음수에도 그대로 작동한다.
- 주의할 부분은 손실함수보다 출력층이다. sigmoid/ReLU처럼 음수 출력을 막는 activation을 쓰면 기온 회귀에 부적절할 수 있다.

### 최종 답안형

기온은 섭씨 기준으로 음수도 가능한 연속 수치형 target이다. 따라서 기본적으로 회귀 문제다. 섭씨온도는 `0`이 절대적 없음의 의미가 아니므로 비율척도라기보다 등간척도에 가깝다. MSE는 음수 target에도 그대로 사용할 수 있으므로 손실함수 수식 자체를 수정할 필요는 없다. 다만 출력층은 음수 예측이 가능하도록 linear output을 사용해야 하며, sigmoid나 ReLU처럼 출력 범위를 제한하는 activation은 부적절할 수 있다. 분류로 바꾸려면 `매우추움/추움/보통/더움/매우더움`처럼 구간을 정의하고, 이 경우 순서형 분류가 된다.

## 마지막 정리

헷갈림은 `회귀 vs 분류` 하나가 아니라 아래 세 축이 섞일 때 생긴다.

| 축 | 질문 | 예 |
|---|---|---|
| 수치량인가? | 실제 양/크기/단위가 있는가? | 점수, 집값, 기온, 혈당 |
| class encoding인가? | 이름을 숫자로 바꾼 것인가? | 성별 `0/1`, 장르 `0/1/2/3`, Iris `0/1/2` |
| 순서형인가? | class 사이에 순서는 있는가? | 만족도 `1~5`, 안전등급 `0~3`, 혼잡도 `0~3` |

문제를 풀 때는 항상 `y`에 대해 먼저 묻는다.

1. 이 숫자가 실제 양/크기/단위인가?
2. 아니면 class 이름을 숫자로 바꾼 것인가?
3. class라면 순서가 있는가?
4. 순서가 있다면 간격까지 믿을 수 있는가?
