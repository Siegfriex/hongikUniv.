## C-3. 영화 장르 encoding

### 문제

영화 장르를 다음처럼 저장했다.

- 0 = 액션
- 1 = 로맨스
- 2 = 스릴러
- 3 = 다큐멘터리

줄거리 텍스트를 이용해 장르 번호를 예측한다.

### 답할 것

- X =
- y =
- feature =
- target =
- 이 숫자는 크기/거리 의미가 있는가?
- class 이름을 숫자로 바꾼 것인가?
- 문제 유형 =
- 회귀로 풀면 생기는 문제 =

````python
import pandas as pd
import numpy as np
import re

target_names = np.array(["액션", "로맨스", "스릴러", "다큐멘터리"])
raw_df = pd.DataFrame({
    "plot": [
        "특수부대 요원이 폭발과 총격 액션 속에서 테러리스트를 추격한다.",
        "두 사람이 우연히 만나 사랑과 로맨스를 키우지만 이별 위기에 놓인다.",
        "연쇄 살인범을 쫓는 형사가 어두운 도시에서 스릴러 같은 긴장을 겪는다.",
        "북극곰과 기후 변화를 담은 다큐멘터리 영화가 자연의 아름다움을 보여준다.",
    ],
    "genre_code": [0, 1, 2, 3],
})
keywords = ["액션", "로맨스", "스릴러", "다큐멘터리"]

def extract_keyword_features(text, keywords):
     words = re.findall(r"[가-힣]+", text)
     total_words = max(len(words), 1)
     result = {}
     for kw in keywords:
        count = text.count(kw)
        result[f"{kw}_exists"] = int(count > 0)
        result[f"{kw}_count"] = count
        result[f"{kw}_ratio"] = count / total_words
     return result

feature_df = raw_df["plot"].apply(lambda t: pd.Series(extract_keyword_features(t, keywords)))

X_df = feature_df[[f"{kw}_ratio" for kw in keywords]].copy()
X_df.columns = [f"feature_{i+1}_{kw}" for i, kw in enumerate(keywords)]
X = X_df.values
y = raw_df["genre_code"].values
y_s = pd.Series(y, name="genre_code")
print(X_df)
print(feature_df)
print("target 이름:", target_names)
print("y 고유값:", np.unique(y))
print("클래스별 개수:")
print(y_s.value_counts().sort_index())

print("=== 로데이터 (줄거리 + 장르 code) ===")
print(raw_df.head())



````

````python
import pandas as pd
import numpy as np

target_names = ["원활", "보통", "혼잡", "매우혼잡"]

#피처1 : 차량수 / 도로용량 비율
#피처2 : 시간대별 가중치/요일별 가중치 
#피처3 : 평균속도/도로구간 비율
#피처4 : 날씨별 가중치 
#피처5 : 사고여부/도로구간 비율
#피처6 : 차량수 / 시간대별 기대용량 비율

## 피처 정의표 (Gate 답안용)
feature_spec = pd.DataFrame({
    "피처": ["feature_1", "feature_2", "feature_3", "feature_4", "feature_5", "feature_6"],
    "한글명": [
        "차량수/도로용량 비율",
        "시간대가중치/요일가중치",
        "평균속도/도로길이 비율",
        "날씨 가중치",
        "사고여부/도로길이 비율",
        "차량수/시간대용량 비율",
    ],
    "수식": [
        "vehicle_count / road_capacity",
        "time_weight[hour] / day_weight[dayofweek]",
        "avg_speed / road_km",
        "weather_weight[weather]",
        "accident / road_km",
        "vehicle_count / hour_capacity[hour]",
    ],
    "타입": ["비율", "비율", "비율", "가중치", "비율", "비율"],
})


# 보조 매핑 (로데이터를 사람이 읽게)
day_names = ["월", "화", "수", "목", "금", "토", "일"]
hour_bucket = {
    range(0, 6): "새벽",
    range(6, 10): "출근",
    range(10, 17): "낮",
    range(17, 20): "퇴근",
    range(20, 24): "야간",
}
def get_hour_bucket(h):
    for r, name in hour_bucket.items():
        if h in r:
            return name
    return "기타"

# 1) 로데이터 직접 제작
raw_df = pd.DataFrame({
    "vehicle_count": [80, 150, 320, 410, 200, 90, 380, 420],
    "hour":          [3, 10, 8, 18, 14, 6, 19, 8],
    "dayofweek":     [1, 2, 1, 4, 3, 6, 5, 1],
    "weather":       ["맑음", "맑음", "비", "흐림", "맑음", "맑음", "비", "눈"],
    "avg_speed":     [70, 55, 25, 15, 45, 65, 20, 10],
    "road_km":       [3.0, 5.0, 4.0, 6.0, 5.0, 3.0, 4.0, 5.0],
    "road_capacity": [200, 300, 400, 400, 300, 200, 400, 400],
    "accident":      [0, 0, 1, 1, 0, 0, 1, 1],
    "congestion_code": [0, 1, 2, 3, 1, 0, 2, 3],
})


time_weight = {h: 1.0 for h in range(24)}
for h, w in {7:1.5, 8:2.0, 9:1.8, 17:1.9, 18:2.2, 19:1.7}.items():
    time_weight[h] = w

day_weight = {0:1.0, 1:1.0, 2:1.0, 3:1.0, 4:1.0, 5:0.7, 6:0.6}

weather_weight = {"맑음":1.0, "흐림":1.1, "비":1.3, "눈":1.5}

hour_capacity = {h: (300 if h in [7,8,17,18,19] else 200) for h in range(24)}

def extract_traffic_features(row):
    return pd.Series({
        "feature_1_vehicle_capacity_ratio": row["vehicle_count"] / row["road_capacity"],
        "feature_2_time_day_weight_ratio": time_weight[row["hour"]] / day_weight[row["dayofweek"]],
        "feature_3_speed_road_ratio": row["avg_speed"] / row["road_km"],
        "feature_4_weather_weight": weather_weight[row["weather"]],
        "feature_5_accident_road_ratio": row["accident"] / row["road_km"],
        "feature_6_vehicle_hour_ratio": row["vehicle_count"] / hour_capacity[row["hour"]],
    })


# 2) 피처 추출
feature_df = raw_df.apply(extract_traffic_features, axis=1)
X_df = feature_df.copy()
X = X_df.values
y = raw_df["congestion_code"].values
y_s = pd.Series(y, name="congestion_code")

# 통합 테이블: 로데이터 + 해석 컬럼 + 피처 + 정답
df = raw_df.copy()
df["요일"] = df["dayofweek"].map(lambda d: day_names[d])
df["시간대구간"] = df["hour"].map(get_hour_bucket)
df["혼잡도"] = df["congestion_code"].map(dict(enumerate(target_names)))
df = pd.concat([df, X_df], axis=1)

# 클래스 코드 → 한글 라벨 매핑 (rename(index=배열)은 Series에서 오류)
class_counts = y_s.value_counts().sort_index()
class_counts.index = [target_names[i] for i in class_counts.index]

print("=" * 60)
print("【1】 피처 정의")
print(feature_spec.to_string(index=False))

print("\n" + "=" * 60)
print("【2】 로데이터 + 분류 라벨")
print(df[[
    "vehicle_count", "hour", "시간대구간", "dayofweek", "요일",
    "weather", "avg_speed", "road_km", "accident",
    "congestion_code", "혼잡도"
]].to_string(index=True))

print("\n" + "=" * 60)
print("【3】 X / y 크기")
print(f"X shape: {X.shape}  → {X.shape[0]}개 샘플, {X.shape[1]}개 피처")
print(f"y shape: {y.shape}  → 클래스 코드 {list(np.unique(y))}")
print(f"target 이름: {list(target_names)}")

print("\n" + "=" * 60)
print("【4】 클래스별 개수")
print(class_counts)

print("\n" + "=" * 60)
print("【5】 피처 통계")
print(X_df.describe().round(3).to_string())

print("\n" + "=" * 60)
print("【6】 샘플별 한 줄 요약")
for i, row in df.iterrows():
    print(
        f"[{i}] {row['혼잡도']}({row['congestion_code']}) | "
        f"{row['hour']}시({row['시간대구간']}) {row['요일']} | "
        f"차량 {row['vehicle_count']}대 | 속도 {row['avg_speed']}km/h | "
        f"날씨 {row['weather']} | "
        f"f1={row['feature_1_vehicle_capacity_ratio']:.3f}, "
        f"f6={row['feature_6_vehicle_hour_ratio']:.3f}"
    )

print("\n" + "=" * 60)
print("【7】 Gate 판단 요약")
print("X =", list(X_df.columns))
print("y = congestion_code (0~3)")
print("feature = 위 6개 수치 피처 (비율/가중치)")
print("target =", list(target_names), "← class 이름을 숫자로 인코딩")
print("문제 유형 = 순서형 다중분류 (원활 < 보통 < 혼잡 < 매우혼잡)")
print("회귀로 풀면 = 2.3 같은 중간값 예측, 클래스 경계 모호")

df
````

## C-1. Iris 품종 encoding

### 네 답안 요약

```text
X = shape(150,4)
y = shape(150,)
feature = sepal/petal 네 개
target = setosa, versicolor, virginica
이 숫자는 크기/거리 의미가 있는가? = X 피처는 연속형 수치...
class 이름을 숫자로 바꾼 것인가? = 클래스 이름을 숫자로 바꾼 게 아님
문제 유형 = 다중분류
```

### 채점

| 항목                | 판정      |
| ----------------- | ------- |
| X                 | 부분 정답   |
| y                 | 부분 정답   |
| feature           | 정답      |
| target            | 정답에 가까움 |
| 숫자 의미 판단          | 오답      |
| class encoding 판단 | 오답      |
| 문제 유형             | 정답      |
| 회귀로 풀면 생기는 문제     | 불충분     |

### 구체 교정

`X = shape(150,4)`라고 쓴 것은 틀리지는 않지만 부족하다. (X)는 단순히 shape가 아니라 **꽃의 네 가지 측정값 전체**다.

더 정확한 표현은 이렇다.

```text
X = sepal length, sepal width, petal length, petal width로 구성된 입력 행렬
X.shape = (150, 4)
```

`y = shape(150,)`도 마찬가지다. shape만 쓰면 target의 의미가 빠진다.

```text
y = species_code, 즉 setosa/versicolor/virginica를 0/1/2로 인코딩한 값
y.shape = (150,)
```

가장 중요한 오답은 여기다.

```text
class 이름을 숫자로 바꾼 것인가? = 클래스 이름을 숫자로 바꾼 게 아님
```

이건 반대로 써야 한다. Iris에서 `0=setosa`, `1=versicolor`, `2=virginica`는 **class 이름을 숫자로 바꾼 것**이다. `0,1,2` 사이에는 크기나 거리 의미가 없다. `virginica`가 `setosa`보다 “2만큼 큰 품종”이라는 뜻이 아니다.

회귀로 풀면 생기는 문제도 더 명확해야 한다. 회귀로 풀면 모델은 `0,1,2` 사이의 거리와 순서가 의미 있다고 가정한다. 예를 들어 예측값이 `1.7`이면 이것을 어떻게 해석할지 애매해진다. 또한 MSE는 `0→2` 오류를 `0→1` 오류보다 더 큰 수치 오차로 취급하는데, 품종 class에 그런 거리 의미가 있다는 보장은 없다.

### C-1 점수

**60/100.**
X/y와 문제 유형은 맞았지만, Drill C의 핵심인 class encoding 판단을 반대로 썼다.

## C-2. 고객 만족도 1~5점

### 네 답안 요약

```text
feature = 응답 시간, 재방문 횟수, 불만 키워드 수, 구매 금액, 상담 횟수

문제 유형 = 5중 분류
다른 방식 = 회귀 후 반올림 가능
```

### 채점

| 항목                | 판정      |
| ----------------- | ------- |
| X                 | 누락      |
| y                 | 누락에 가까움 |
| feature           | 정답      |
| target            | 부분 정답   |
| 숫자 의미 판단          | 부분 오답   |
| class encoding 판단 | 부분 정답   |
| 문제 유형             | 부분 정답   |
| 다른 방식 설명          | 괜찮음     |

### 구체 교정

C-2는 애매한 문제다. 이 문제를 잘 푸는 방식은 “하나로 단정”이 아니라 **왜 애매한지**를 설명하는 것이다.

고객 만족도 `1,2,3,4,5`는 Iris의 `0,1,2`와 다르다. Iris 숫자는 완전히 명목형 class encoding이다. 하지만 만족도 점수는 보통 다음 순서를 가진다.

```text
1 < 2 < 3 < 4 < 5
```

즉, **순서 의미는 있다.**
하지만 `1점→2점` 차이와 `4점→5점` 차이가 정확히 같은 심리적 거리인지 확실하지 않다. 그래서 이것은 보통 **순서형 target**이다.

가능한 접근은 세 가지다.

```text
1. 다중분류: 1,2,3,4,5를 다섯 class로 본다.
2. 순서형 분류: 1<2<3<4<5 순서를 반영한다.
3. 회귀: 1~5를 거의 연속 점수처럼 보고 예측 후 반올림한다.
```

네가 “5중 분류”라고 한 것은 완전히 틀리지는 않다. 하지만 가장 좋은 답은 **순서형 분류 또는 ordinal target**이라고 쓰는 것이다.

“이 숫자는 크기/거리 의미가 있는가?”에 대한 네 답은 질문을 잘못 잡았다. 여기서 묻는 숫자는 feature의 단위가 아니라 **y = 1~5 만족도 점수**다.

더 정확히는:

```text
이 숫자는 순서 의미는 있다.
다만 정확한 거리 의미는 약하다.
```

“class 이름을 숫자로 바꾼 것인가?”도 애매하다. 만약 실제 설문이 `매우불만족/불만족/보통/만족/매우만족`을 1~5로 저장한 것이라면 class 이름을 숫자로 매핑한 것이다. 다만 완전한 명목형 class encoding이 아니라 **순서형 encoding**이다.

### C-2 점수

**70/100.**
문제 유형을 분류로 잡은 것은 가능하지만, “순서형”이라는 핵심 표현이 빠졌고 숫자 의미 판단이 흔들렸다.

## C-3. 영화 장르 encoding

### 네 답안 요약

```text
X = 액션/로맨스/스릴러/다큐멘터리의 exists/count/ratio
y = 0 1 2 3
feature = 12개
target = 분류, 0 1 2 3
이 숫자는 크기/거리 의미가 있는가? = exists/count/ratio 설명
class 이름을 숫자로 바꾼 것인가? = 공란
문제 유형 = 분류
회귀로 풀면 생기는 문제 = 연속형 출력, integer가 아니면 결정계수로 정해야 함
```

### 채점

| 항목                | 판정    |
| ----------------- | ----- |
| X                 | 부분 정답 |
| y                 | 정답    |
| feature           | 부분 정답 |
| target            | 부분 정답 |
| 숫자 의미 판단          | 오답    |
| class encoding 판단 | 누락    |
| 문제 유형             | 정답    |
| 회귀 문제 설명          | 부분 정답 |

### 구체 교정

코드 자체는 좋은 시도다. 텍스트를 그대로 모델에 넣지 않고 `exists`, `count`, `ratio` 같은 수치 feature로 바꾼 것은 Gate 0 관점에서 적절하다.

다만 네 코드에서는 실제 (X)로 무엇을 썼는지 답안이 조금 불일치한다. 코드에서는 최종 `X_df`를 ratio 4개로 만들었다.

```text
feature_1_액션
feature_2_로맨스
feature_3_스릴러
feature_4_다큐멘터리
```

그런데 답안에는 exists/count/ratio 12개라고 썼다. 둘 중 하나로 고정해야 한다.

가능한 답안은 둘 중 하나다.

```text
버전 A:
X = 네 장르 키워드 ratio 4개
feature = 액션_ratio, 로맨스_ratio, 스릴러_ratio, 다큐멘터리_ratio
```

또는

```text
버전 B:
X = 네 장르 키워드의 exists/count/ratio 12개
feature = 액션_exists, 액션_count, 액션_ratio, ... 총 12개
```

이번 문제의 핵심 오답은 또 같은 지점이다.

```text
이 숫자는 크기/거리 의미가 있는가?
```

여기서 “이 숫자”는 feature의 exists/count/ratio가 아니라 **y = 0,1,2,3 장르 코드**다.

정확한 판단은 다음이다.

```text
0,1,2,3은 크기/거리 의미가 없다.
0=액션, 1=로맨스, 2=스릴러, 3=다큐멘터리라는 class 이름을 숫자로 바꾼 것이다.
```

영화 장르에 순서나 거리 의미는 없다. 액션이 로맨스보다 작거나, 다큐멘터리가 스릴러보다 큰 것이 아니다. 따라서 이것은 **명목형 다중분류**다.

회귀로 풀면 생기는 문제는 다음처럼 써야 한다.

```text
회귀로 풀면 장르 코드 사이에 순서와 거리 의미가 있다고 가정한다.
예측값이 1.6처럼 나오면 어떤 장르인지 애매하다.
또한 MSE가 액션(0)과 다큐(3)의 차이를 액션(0)과 로맨스(1)의 차이보다 더 큰 오류로 처리하는데,
장르 class에는 그런 수치적 거리 의미가 없다.
```

“결정계수”라는 표현은 여기서 부적절하다. 결정계수 (R^2)는 회귀 평가 지표이고, class를 정하는 기준이 아니다. 여기서는 “threshold” 또는 “rounding”이라고 해야 한다.

### C-3 점수

**68/100.**
텍스트 feature engineering 시도는 좋지만, target code의 의미 판단이 빠졌고 회귀 문제 설명에 용어 오류가 있다.
