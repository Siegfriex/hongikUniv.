# Gate 0-1 보완문제 정답본

이 노트북은 보완문제 채점표를 바탕으로 만든 정답/복습용 노트북이다.

- 본문은 정답 기준으로 작성했다.
- 네가 이전 답변에서 잘못 잡았거나 불충분했던 지점은 각 문항의 `이전 오답 주석`에 명시했다.
- 보완 4, 보완 5는 미응답으로 채점 제외였으므로 포함하지 않았다.

## 공통 판단 기준

6강 기준:

- `X` = 예측변수, feature, 입력 변수
- `y` = 반응변수, target, 예측 대상
- 분류 = 범주 label 예측
- 회귀 = 실제 수치값 예측

12강 기준:

- 분류 = softmax, cross entropy, one-hot target 등과 연결
- 회귀 = linear output, MSE, 실수값 target 등과 연결

15강 기준:

- 순서 없는 범주 = 명목형 class encoding
- 순서 있는 범주 = ordinal target 또는 ordinal encoding
- ordinal encoding은 모델이 숫자 간격까지 의미 있다고 받아들일 수 있으므로 주의해야 한다.


## 보완 1. 성별 0=남성, 1=여성

### 정답

**X** = 성별을 예측하기 위한 입력 feature들  
예: 키, 체중, 음성 특징, 설문 응답 등 문제에서 주어진 예측변수

**y** = 성별 class

**target** =

- `0 = 남성`
- `1 = 여성`

**숫자 의미**  
`0`과 `1`은 크기나 거리 의미가 없다. 남성이 여성보다 작거나, 여성이 남성보다 1만큼 큰 것이 아니다.

**class encoding 여부**  
성별이라는 명목형 class를 `0/1`로 인코딩한 것이다.

**문제 유형**  
명목형 이진분류 문제다. 표현은 `binary classification` 또는 `Bernoulli target`이 적절하다.

**회귀로 풀면 생기는 문제**  
회귀로 풀면 `0.4`, `0.7` 같은 연속 출력이 나올 수 있다. 이 값은 “성별의 중간값”이라는 실제 수치 의미를 갖지 않는다. 분류 모델에서는 이런 값을 class 1일 확률로 해석하고 threshold로 최종 class를 정한다.

### 이전 오답 주석

- 전체 방향은 거의 맞았다.
- `바이노미널`보다는 `이진분류(binary classification)` 또는 `Bernoulli target`이 더 안전한 표현이다.
- 로지스틱 회귀나 sigmoid classifier의 `0.7`은 성별의 회귀값이 아니라 class 1의 확률이라는 점을 더 명확히 해야 한다.

### 최종 답안형

`0/1`은 성별 class를 숫자로 인코딩한 것이며 크기나 거리 의미가 없다. 따라서 이 문제는 명목형 이진분류다. 회귀로 풀면 `0.4`, `0.7` 같은 연속 출력이 나오며, 이는 성별이라는 class의 실제 수치값으로 해석될 수 없다. 분류 모델에서는 이 값을 class 1의 확률로 해석하고 threshold로 최종 class를 정한다.


## 보완 2. 안전등급 0=정상, 1=주의, 2=위험, 3=매우위험

### 정답

**X** = 안전등급을 예측하기 위한 입력 feature들  
예: 센서값, 이상 징후 수, 온도, 진동, 압력, 사고 이력 등

**y** = 안전등급 class

**target** =

- `0 = 정상`
- `1 = 주의`
- `2 = 위험`
- `3 = 매우위험`

**숫자 의미**  
순서 의미는 있다. `정상 < 주의 < 위험 < 매우위험`이라는 위험도 방향이 존재한다.

하지만 거리/간격 의미는 불명확하다. `정상 -> 주의`의 차이와 `위험 -> 매우위험`의 차이가 실제 위험도에서 정확히 같은 간격이라고 보장할 수 없다.

**class encoding 여부**  
class 이름을 숫자로 인코딩한 것이다. 다만 순서가 있는 class encoding이다.

**문제 유형**  
순서형 분류, 즉 ordinal classification이 가장 자연스럽다.

**회귀로 풀면 생기는 문제**  
일반 회귀로 풀면 `0,1,2,3` 사이의 간격이 실제 위험도 차이와 동일하다고 가정하게 된다. 이 가정은 강하다. 대신 정상에서 매우위험으로 틀리는 오류를 더 크게 벌주는 ordinal loss나 cost-sensitive classification은 가능하다.

### 이전 오답 주석

- `0 < 1 < 2 < 3`의 순서 의미와 간격 불명확성을 잡은 것은 좋았다.
- 다만 “분류된 것에 대해 가중을 매길 수 있다”는 말은 모델링 설계 선택으로 정리해야 한다.
- 가중/비용을 줄 수 있다는 사실이 곧 `0,1,2,3`을 단순 회귀 target처럼 써도 된다는 뜻은 아니다.

### 최종 답안형

안전등급 `0/1/2/3`은 class encoding이다. 다만 `정상 < 주의 < 위험 < 매우위험`이라는 순서가 있으므로 순서형 분류로 보는 것이 적절하다. 숫자 간격이 실제 위험도 차이와 동일하다고 보장되지 않으므로 일반 회귀로 풀면 거리 의미를 과하게 가정하게 된다. 다만 심각한 오분류에 더 큰 비용을 주는 ordinal loss나 cost-sensitive classification은 가능하다.


## 보완 3. 수학점수 0~100

### 정답

**X** = 수학점수를 예측하기 위한 입력 feature들  
예: 공부 시간, 출석률, 과제 점수, 이전 시험 점수, 문제 풀이량 등

**y** = 수학점수 `0~100`

**숫자 의미**  
실제 수치량이다. `80점`과 `90점`의 차이는 10점이고, `40점`과 `50점`의 차이도 10점이다.

**class encoding 여부**  
아니다. `0~100`은 class 이름을 숫자로 바꾼 것이 아니라 실제 측정된 점수값이다.

**문제 유형**  
기본적으로 회귀 문제다.

**등급화하면 어떻게 되는가?**

원래 점수 `0~100` 자체는 ordinal classification이 아니다. 회귀 target이다.

하지만 점수를 구간화하면 문제 유형이 바뀐다.

- 원점수 `0~100` 예측 = 회귀
- `90점 이상 A`, `80점 이상 B` 등으로 변환 = 순서형 다중분류
- `1~9등급`으로 변환 = 순서형 분류
- `합격/불합격`으로 변환 = 이진분류

등급으로 바꾼 순간 모델의 target은 더 이상 정확한 점수값이 아니라 구간 class다. 이 경우 class 사이에 순서는 있지만, 등급 코드 간 간격이 원래 점수 차이를 완전히 보존한다고 보기는 어렵다.

### 이전 오답 주석

- `수학점수 0~100 = 실제 수치량 = 회귀`까지는 맞았다.
- “순서형 인코딩으로서 2와 달리 거리가 의미가 있다”는 표현은 교정이 필요하다.
- 원점수는 거리 의미가 있지만, 점수를 등급으로 바꾼 순간 target은 구간 class가 된다.
- 특히 9등급제가 percentile 기반이면 등급 간 점수 폭이 동일하지 않을 수 있다.

### 최종 답안형

수학점수 `0~100`은 실제 수치량이므로 기본적으로 회귀 문제다. 분류로 바꾸려면 점수 구간을 미리 정의해야 한다. 예를 들어 `90점 이상 A`, `80점 이상 B`처럼 바꾸면 `y`는 순서형 class가 된다. 이 경우 class 사이에 순서는 있지만, 등급 코드 간 간격이 원래 점수 차이를 완전히 보존한다고 보기는 어렵다. 따라서 원점수를 예측하려면 회귀, 등급을 예측하려면 순서형 분류가 적절하다.


## 보완 6. 기온 예측

### 정답

**X** = 기온을 예측하기 위한 입력 feature들  
예: 날짜, 시간, 지역, 습도, 풍속, 기압, 계절, 과거 기온 등

**y** = 기온 값  
예: 섭씨 `-5°C`, `0°C`, `23.5°C`, `31°C`

**숫자 의미**  
기온은 실제 수치값이다. 음수도 가능하며, 값의 차이는 온도 차이를 의미한다.

**class encoding 여부**  
아니다. 기온은 class 이름을 숫자로 바꾼 것이 아니라 연속 수치형 target이다.

**문제 유형**  
기본적으로 회귀 문제다.

**척도 해석**

섭씨온도는 보통 등간척도에 가깝다. 섭씨 `0°C`는 물이 어는 기준점이지 “온도가 없음”을 뜻하지 않는다. 따라서 섭씨는 음수도 가능하다.

- 섭씨온도 = 등간척도에 가까움
- 켈빈온도 = 절대 0이 있으므로 비율척도에 가까움

**MSE와 음수 target**

MSE는 음수 target에도 그대로 사용할 수 있다. 수식을 수정할 필요가 없다.

예를 들어 실제 기온이 `-5°C`, 예측 기온이 `-3°C`라면 오차는 `-2`이고 제곱오차는 `4`다. 정상적으로 계산된다.

**출력층 주의**

수정해야 할 것은 loss가 아니라 output activation이다. 기온은 음수 예측이 가능해야 하므로 회귀 모델의 마지막 출력층은 보통 linear output, 즉 activation 없음이 적절하다. sigmoid나 ReLU처럼 출력 범위를 제한하는 activation은 부적절할 수 있다.

**분류로 바꾸는 경우**

기온을 `매우추움/추움/보통/더움/매우더움`처럼 구간화하면 순서형 분류 문제가 된다. 하지만 이때는 원래의 정확한 기온값을 예측하는 것이 아니라 온도 구간 class를 예측하는 것이다.

### 이전 오답 주석

- `기온은 실제 수치값`, `음수 가능`, `기본적으로 회귀`, `분류로 바꾸려면 급간화 가능`은 맞았다.
- `섭씨온도는 절대영도 기준`이라는 설명은 틀렸다. 섭씨 `0°C`는 절대적 0이 아니다.
- `수치형 범주`라는 표현은 부적절하다. 기온은 범주형이 아니라 연속 수치형 변수다.
- `비율척도 등간척도가 아니다`가 아니라, 섭씨온도는 등간척도에 가깝고 켈빈온도는 비율척도에 가깝다.
- 음수 target 때문에 MSE 수식을 수정해야 하는 것이 아니다. MSE는 음수에도 그대로 작동한다.
- 주의할 부분은 손실함수보다 출력층이다. sigmoid/ReLU처럼 음수 출력을 막는 activation을 쓰면 기온 회귀에 부적절할 수 있다.

### 최종 답안형

기온은 섭씨 기준으로 음수도 가능한 연속 수치형 target이다. 따라서 기본적으로 회귀 문제다. 섭씨온도는 `0`이 절대적 없음의 의미가 아니므로 비율척도라기보다 등간척도에 가깝다. MSE는 음수 target에도 그대로 사용할 수 있으므로 손실함수 수식 자체를 수정할 필요는 없다. 다만 출력층은 음수 예측이 가능하도록 linear output을 사용해야 하며, sigmoid나 ReLU처럼 출력 범위를 제한하는 activation은 부적절할 수 있다. 분류로 바꾸려면 `매우추움/추움/보통/더움/매우더움`처럼 구간을 정의하고, 이 경우 순서형 분류가 된다.


## 최종 기준표

| y 예시 | 숫자 의미 | 순서 | 거리/간격 | 문제 유형 |
|---|---|---|---|---|
| 성별 `0/1` | class encoding | 없음 | 없음 | 이진분류 |
| Iris `0/1/2` | class encoding | 없음 | 없음 | 다중분류 |
| 영화 장르 `0/1/2/3` | class encoding | 없음 | 없음 | 다중분류 |
| 안전등급 `0/1/2/3` | class encoding | 있음 | 불명확 | 순서형 분류 |
| 만족도 `1~5` | class/score 혼합 | 있음 | 애매함 | 순서형 분류 또는 회귀 |
| 수학점수 `0~100` | 실제 수치량 | 있음 | 있음 | 회귀 |
| 기온 `°C` | 실제 수치량 | 있음 | 있음 | 회귀 |
| 혈당 `mg/dL` | 실제 수치량 | 있음 | 있음 | 회귀 |
| 집값 `원` | 실제 수치량 | 있음 | 있음 | 회귀 |

## 마지막 정리

헷갈림은 `회귀 vs 분류` 하나가 아니라 아래 세 축이 섞일 때 생긴다.

| 축 | 질문 | 예 |
|---|---|---|
| 수치량인가? | 실제 양/크기/단위가 있는가? | 점수, 집값, 기온, 혈당 |
| class encoding인가? | 이름을 숫자로 바꾼 것인가? | 성별 `0/1`, 장르 `0/1/2/3`, Iris `0/1/2` |
| 순서형인가? | class 사이에 순서는 있는가? | 만족도 `1~5`, 안전등급 `0~3`, 혼잡도 `0~3` |

문제를 풀 때는 항상 `y`에 대해 먼저 묻는다.

1. 이 숫자가 실제 양/크기/단위인가?
2. 아니면 class 이름을 숫자로 바꾼 것인가?
3. class라면 순서가 있는가?
4. 순서가 있다면 간격까지 믿을 수 있는가?
