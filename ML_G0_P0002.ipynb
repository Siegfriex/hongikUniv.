### 내 답안 A-1 (구조 시각화 표 중심 정리)

| 구분         | 주요 객체         | 역할/설명                                                        | 사람이 보기 좋은가 | 모델이 계산하기 좋은가 | 주요 정보/특징 예시                                                         |
|:-------------|:----------------:|:------------------------------------------------------------------|:------------------:|:----------------------:|:----------------------------------------------------------------------------|
| **1. DataFrame** | ![DataFrame](https://img.shields.io/badge/pandas-DataFrame-lightgrey) | 표 형태(2D), 컬럼명/인덱스/행렬, EDA·전처리·피처해석에 용이<br>SQL/R의 데이터표 느낌, 다양한 통계/조회 지원 | ✅                  | ❌                    | `.columns`, `.index`, `.info()`, `.describe()`, `.isna()`, `.nunique()` 등  |
| **2. Series**    | ![Series](https://img.shields.io/badge/pandas-Series-blue)            | 1차원, 라벨있는 벡터<br>컬럼 하나만 떼어낸 형태                   | ➖                 | ➖                   | `df['col']` → Series<br>`s.shape=(N,)`                                      |
| **3. ndarray**   | ![ndarray](https://img.shields.io/badge/numpy-ndarray-green)          | 순수 배열, 컬럼·인덱스 정보 없음<br>배치/특성 순서만으로 계산     | ❌                  | ✅                    | `.shape`, `.dtype`, 연산속도 빠름, 피처 순서 주의 필요                      |
| **4. tensor**    | ![Tensor](https://img.shields.io/badge/deep%20lib-tensor-red)         | 다차원 배열(ndarray 확장)<br>GPU/딥러닝용, 자동 미분 등            | ❌                  | ✅✅                   | `.shape`, `.dtype`, batch 연산, GPU 가속, flatten·norm 등 다양한 처리        |

#### 이해를 돕는 이미지 예시

```
        DataFrame (pandas)
┌─────┬────────────┬───────┬──────────┐
│id   │ feature1   │ ...   │ label    │
├─────┼────────────┼───────┼──────────┤
│ ... │  ...       │ ...   │ ...      │
└─────┴────────────┴───────┴──────────┘
[컬럼명, 인덱스 모두 보임]
```
---
```
        ndarray (numpy)
[[1.2, 3.4, ...],  # 피처1,2,...
 [0.5, 6.1, ...],

[값만 있고, 컬럼명·인덱스 없음]
```
---
```
        tensor (예: 이미지)
[ [ [0.8, 0.4, ...], [0.9, ...], ... ],   # (H, W, C) 또는 (batch, ...)
  [ ... ], ... ]
[4차원 가능, 연산은 GPU, 구조만 중요]
```

---

#### 키워드 비교 요약

| 객체      | 사람이 보기 좋은가 | 모델이 계산 좋은가 | 예시                       | 주의할 점/비교포인트                 |
|:----------|:-----------------:|:-------------------:|:--------------------------|:-------------------------------------|
| DataFrame | O                 | X                   | df.head(), df.info()      | 컬럼명/인덱스 추적 가능              |
| Series    | O                 | X                   | df['score']               | 1D, 레이블 있음                      |
| ndarray   | X                 | O                   | X.values, X.to_numpy()    | 순수 배열, 컬럼명 사라짐             |
| tensor    | X                 | O                   | torch.Tensor, tf.Tensor   | GPU 지원, shape/차원 유연            |

---

#### 정리

- **사람이 보기 좋은 구조** : **DataFrame** (표, 구조적, 해석·전처리·탐색)
- **모델이 계산하기 좋은 구조** : **ndarray, tensor** (배열 중심, 벡터화, 연산성능)
- **주의할 점** :
    - DataFrame → 배열(ndarray, tensor)로 바꿀 때 컬럼명·인덱스 정보가 사라져 순서 맞추기 실수 유의
    - tensor는 GPU 연산 가능, 자동미분, 고차원 데이터에 효과적
    - shape 개념(explicit): (N, D) vs (batch, channel, height, width) 등 다양함

````python
# 힌트:
# 이 문제는 코드보다 설명이 중요하다.
# DataFrame = 의미 확인 / EDA / 전처리 설계
# ndarray/tensor = 수치 계산 / 모델 학습
# 변환 시 column name, dtype, 결측치, 인코딩 문제를 생각한다.

import pandas as pd

# =======================
# traffic_congestion.csv 데이터 로딩 및 전처리 단계 (AssertionError 및 디버깅 대응)
# =======================

# [1] 데이터 파일 불러오기 (DATA_DIR 경로상 파일 필요)
traffic = pd.read_csv(DATA_DIR / "traffic_congestion.csv")
display(traffic.head())  # 데이터 shape, 컬럼, 결측 등 확인

# [2] 타깃 컬럼 올바르게 지정 (문제 핵심: target_col 변수값 재확인!)
# 오류 핸들링: target_col이 traffic 컬럼에 실제 존재하지 않으면, 명시적으로 올바른 이름으로 재정의
print("데이터 컬럼 목록:", list(traffic.columns))
# 아래 라인은 교재 예시와 다르게, 사용자가 오류를 낼 수 있으므로 robust하게 처리
if "congestion_level" in traffic.columns:
    target_col = "congestion_level"
else:
    raise ValueError(
        f"[ERROR] 'congestion_level' 컬럼이 traffic CSV에 없습니다. 컬럼 목록: {list(traffic.columns)}"
    )
print("target_col =", target_col)  # 실제 타깃 이름 print
````

````python
# [설명 중심 리라이트: DataFrame <-> ndarray 전환 사고를 강조, 코드 최소화]
# + KeyError 디버깅 반영 (y.dtypes의 key는 0이 아니라 y.columns[0] 사용)

# --- 힌트 및 요약 ---
# - DataFrame: feature 명/의미 보존, EDA, 전처리 설계에 최적 (컬럼명을 통한 해석/탐색)
# - ndarray/tensor: ML/DL 모델에서 빠른 수치계산에 최적, column name 등 정보는 사라짐
# - ⚠️ 변환 시 주의: 컬럼명 소실, dtype 강제화(int/float), 결측치/범주형 인코딩/label 매핑 정보 따로 기록 필요
# - label(예컨대 congestion_level)의 '숫자값'이 의미하는 범주는 DataFrame 단계에서만 명확히 남으므로 반드시 보조자료로 기록!

import pandas as pd
import numpy as np

traffic = pd.read_csv(DATA_DIR / "traffic_congestion.csv")
target_col = "congestion_level"
categorical_features = traffic.select_dtypes(include=["object", "str"]).columns.tolist()
numeric_features = [col for col in traffic.select_dtypes(include=["int64", "float64"]).columns if col != target_col]

# X, y 준비 (DataFrame 환경: 의미/EDA 중심)
X = traffic[categorical_features + numeric_features]
y = traffic[[target_col]]

print("DataFrame X shape:", X.shape)
print("DataFrame y shape:", y.shape)
X_np = pd.get_dummies(X, columns=categorical_features, dtype=int).to_numpy()
y_np = y.to_numpy()
print("ndarray X_np shape:", X_np.shape)
print("ndarray y_np shape:", y_np.shape)
````

````python
# [텐서(tensor) 중심 + keras/tf 적용 + maplit(→ seaborn/plotly 등)으로 시각화]
# 데이터 전처리 후 keras/tensorflow용 tensor로 변환, feature별 통계·분포를 시각화
# ※ maplit 언급은 실수로 보이며, Python 일반 환경에서 seaborn/plotly/matplotlib 등 실제 사용 예시로 대응!

import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

traffic = pd.read_csv(DATA_DIR / "traffic_congestion.csv")
target_col = "congestion_level"
categorical_features = traffic.select_dtypes(include=["object", "str"]).columns.tolist()
numeric_features = [col for col in traffic.select_dtypes(include=["int64", "float64"]).columns if col != target_col]

# --- DataFrame → One-hot 인코딩 → ndarray(float32) → TensorFlow tensor ---
X = traffic[categorical_features + numeric_features]
y = traffic[[target_col]]

X_encoded = pd.get_dummies(X, columns=categorical_features, dtype=int)
X_np = X_encoded.to_numpy(dtype=np.float32)
y_np = y.to_numpy(dtype=np.int64)
X_tensor = tf.convert_to_tensor(X_np, dtype=tf.float32)
y_tensor = tf.convert_to_tensor(y_np, dtype=tf.int64)
print("X_tensor shape:", X_tensor.shape)
print("y_tensor shape:", y_tensor.shape)
````
